```python
nums = [4,1,2,1,2]
```

```python

# this approach has time limit excede

n = len(nums)
s = 0
for i in range(n):
    visited = nums[i]
    count = 0
    for j in range(n):
        if visited == nums[j]:
            count+=1
    if count == 0 or count ==1:
        s = visited
s
```

```python
# use xor fuction
s_num= 0
for num in nums:
    s_num ^= num
s_num
```
