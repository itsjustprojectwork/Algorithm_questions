Remove Outermost Parentheses<br><br>A valid parentheses string is either empty "", "(" + A + ")", or A + B, where A and B are valid parentheses strings, and + represents string concatenation.<br><br>- For example, "", "()", "(())()", and "(()(()))" are all valid parentheses strings.<br><br>A valid parentheses string s is primitive if it is nonempty, and there does not exist a way to split it into s = A + B, with A and B nonempty valid parentheses strings.<br><br>Given a valid parentheses string s, consider its primitive decomposition: s = P1 + P2 + ... + Pk, where Pi are primitive valid parentheses strings.<br><br>Return s after removing the outermost parentheses of every primitive string in the primitive decomposition of s.<br><br>Example 1:<br>Input<br>s = "(()())(())"<br>Output<br>"()()()"<br>Explanation<br>The input string is "(()())(())", with primitive decomposition "(()())" + "(())".<br>After removing outer parentheses of each part, this is "()()" + "()" = "()()()".<br><br>Example 2:<br>Input<br>s = "(()())(())(()(()))"<br>Output<br>"()()()()(())"<br>Explanation<br>The input string is "(()())(())(()(()))", with primitive decomposition "(()())" + "(())" + "(()(()))".<br>After removing outer parentheses of each part, this is "()()" + "()" + "()(())" = "()()()()(())".<br><br>Example 3:<br>Input<br>s = "()()"<br>Output<br>""<br>Explanation<br>The input string is "()()", with primitive decomposition "()" + "()".<br>After removing outer parentheses of each part, this is "" + "" = "".<br><br>Constraints:<br>- 1 <= s.length <= 105<br>- s[i] is either '(' or ')'.<br>- s is a valid parentheses string.

```python
s ="((()())(()()))"
result = ''
keep = 0
for ch in s:
    if ch == '(':
        if keep>0:
            result+=ch
        keep+=1
    elif ch == ")":
        keep -=1
        if keep>0:
            result+=ch
result
        
```
