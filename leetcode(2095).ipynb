2095. Delete the Middle Node of a Linked List
<br>
You are given the head of a linked list. Delete the middle node, and return the head of the modified linked list.
<br>
The middle node of a linked list of size n is the ⌊n / 2⌋th node from the start using 0-based indexing, where ⌊x⌋ denotes the largest integer less than or equal to x.
<br>
For n = 1, 2, 3, 4, and 5, the middle nodes are 0, 1, 1, 2, and 2, respectively.
 <br>

Example 1:<br>

<br>
Input: head = [1,3,4,7,1,2,6]<br>
Output: [1,3,4,1,2,6]<br><br>
Explanation:<br>
The above figure represents the given linked list. The indices of the nodes are written below.<br>
Since n = 7, node 3 with value 7 is the middle node, which is marked in red.<br>
We return the new list after removing this node. <br>

```python
class Node:
    def __init__(self,val = 0,next = None):
        self.val = val
        self.next = next
```

```python
values = [1]
n = len(values)
head = None
tail = None
for i in range(n):
    node = Node(values[i])
    if head is None:
        head = node
        tail = node
    else:
        tail.next = node
        tail = node
    
```

```python
current = head
while current is not None:
    print(current.val)
    current = current.next
```

```python
prev = None
fast = head
slow = head
while fast!=None and fast.next is not None:
    prev = slow
    slow = slow.next
    fast= fast.next.next
    
if prev is not None:
    prev.next= slow.next
else:
    head = prev
```
