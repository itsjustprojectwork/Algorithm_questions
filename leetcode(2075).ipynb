```python
encodedText = "iveo    eed   l te   olc" 
rows = 4
```

```python
l = list(encodedText)
s = []
col = len(l)//rows
letters = 0
for r in range(rows):
    w = []
    for c in range(col):
        w.append(l[letters])
        letters+=1
    s.append(w)

st = ''        
k = rows
for r in range(col):

    for c in range(r,k):
        if c>=col:
            continue
        st+=s[c-r][c]

    k+=1

print(st.rstrip())
```
