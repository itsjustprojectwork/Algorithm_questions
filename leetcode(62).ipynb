62. Unique Paths
<br>
There is a robot on an m x n grid. The robot is initially located at the top-left corner (i.e., grid[0][0]). The robot tries to move to the bottom-right corner (i.e., grid[m - 1][n - 1]). The robot can only move either down or right at any point in time.
<br>
Given the two integers m and n, return the number of possible unique paths that the robot can take to reach the bottom-right corner.
<br>
The test cases are generated so that the answer will be less than or equal to 2 * 109.
<br>
Input: m = 3, n = 7<br>
Output: 28<br>
Example 2:<br>
<br>
Input: m = 3, n = 2<br>
Output: 3<br>
Explanation: From the top-left corner, there are a total of 3 ways to reach the bottom-right corner:<br>
1. Right -> Down -> Down<br>
2. Down -> Down -> Right<br>
3. Down -> Right -> Down<br>

the problem was simple it is solved by permutation and combination.<br>in how many ways we can arrage elements of same kind<br><br>
formula = n! /p!*q!*r...<br>
<br>
here p,q,r are number of same kinf elements
<br>
m = rows: therefore the only Down moves can be m-1<br>
n = columns: therefore teh only Right moves can be n-1

```python
m = 3
n = 3
```

```python
r_move = m -1
d_move = n-1
total = r_move + d_move
```

```python
def fac(n)->int:
    if n==1 or n==0:
        return 1
    return n*fac(n-1)
```

```python
int(fac(total) / (fac(r_move)*fac(d_move)))
```
