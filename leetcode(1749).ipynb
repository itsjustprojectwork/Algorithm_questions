1749. Maximum Absolute Sum of Any Subarray<br><br>You are given an integer array nums. The absolute sum of a subarray [numsl, numsl+1, ..., numsr-1, numsr] is abs(numsl + numsl+1 + ... + numsr-1 + numsr).<br><br>Return the maximum absolute sum of any (possibly empty) subarray of nums.<br><br>Note that abs(x) is defined as follows:<br><br>- If x is a negative integer, then abs(x) = -x.<br>- If x is a non-negative integer, then abs(x) = x.<br><br>Example 1:<br>Input<br>nums = [1,-3,2,3,-4]<br>Output<br>5<br>Explanation<br>The subarray [2,3] has absolute sum = abs(2+3) = abs(5) = 5.<br><br>Example 2:<br>Input<br>nums = [2,-5,1,-4,3,-2]<br>Output<br>8<br>Explanation<br>The subarray [-5,1,-4] has absolute sum = abs(-5+1-4) = abs(-8) = 8.<br><br>Constraints:<br>- 1 <= nums.length <= 105<br>- -104 <= nums[i] <= 104

```python
nums = [2,-5,1,-4,3,-2]
current_max = 0
current_min= 0
max_sub = float('-inf')
min_sub = float('inf')

for num in nums:
    
    # max_sub
    current_max+=num
    
    if current_max>num:
        current_max = current_max
    else:
        current_max = num
    
    if current_max>max_sub:
        max_sub = current_max
    
    # min_sub
    current_min+=num
    
    if current_min<num:
        current_min = current_min
    else:
        current_min = num
    if current_min<min_sub:
        min_sub = current_min
    
max(max_sub,abs(min_sub))
```
