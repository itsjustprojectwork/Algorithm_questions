49. Group Anagrams
<br>
Given an array of strings strs, group the anagrams together. You can return the answer in any order.
<br>
 

Example 1:<br>

Input: strs = ["eat","tea","tan","ate","nat","bat"]<br>

Output: [["bat"],["nat","tan"],["ate","eat","tea"]]<br>
<br>
Explanation:
<br>
There is no string in strs that can be rearranged to form "bat".<br>
The strings "nat" and "tan" are anagrams as they can be rearranged to form each other.<br>
The strings "ate", "eat", and "tea" are anagrams as they can be rearranged to form each other.<br>
Example 2:<br>
<br>
Input: strs = [""]<br>

Output: [[""]]<br>

Example 3:<br>

Input: strs = ["a"]<br>

Output: [["a"]]<br>

```python
s = "xyzzzzzzzzzzzz"
t = "zzzzzzzzzzzzyx"

    
```

```python
def anagram(s,t):
    map_s = {}
    map_t = {}
    n_s = len(s)
    n_t = len(t)
    if n_s != n_t:
        return False
    else:
        # mapping the letters
        for i in range(n_s):
            if s[i] in map_s:
                map_s[s[i]] += 1
            else:
                map_s[s[i]] = 1
            if t[i] in map_t:
                map_t[t[i]]+=1
            else:
                map_t[t[i]] =1
        
        # checking is anagram or no
        
        for ch in s:
            if (ch in map_s) and (ch in map_t):
                if map_s[ch]!=map_t[ch]:
                    return False
            else:
                return False
        return True
            
```

```python
# approach 1 tle 
strs = ["xyzzzzzzzzzzzz","zzzzzzzzzzzzyx"]
n = len(strs)
seen = set()
result = []
for i in range(n):
    if strs[i] in seen:
        continue
    r = [strs[i]]
    seen.add(strs[i])
    for j in range(i+1,n):
        if anagram(strs[i],strs[j]):
            r.append(strs[j])
            seen.add(strs[j])
    result.append(r)

result
```
