196. Delete Duplicate Emails

```python
import pandas as pd

person = pd.DataFrame({
    "id": [3, 2, 1],
    "email": [
        "john@example.com",
        "bob@example.com",
        "john@example.com"
    ]
})

print(person)
```

```python
person.sort_values('id',inplace = True) # inplace =True change the original dataframe
person.drop_duplicates('email',inplace = True)
```
