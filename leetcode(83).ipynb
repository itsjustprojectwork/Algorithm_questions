```python
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next
```

```python
head = ListNode(1)
head.next = ListNode(1)
head.next.next = ListNode(2)
head.next.next.next = ListNode(3)
head.next.next.next.next = ListNode(3)
```

```python
print('original node:\n')
original_curr = head
while original_curr:
    print(original_curr.val)
    original_curr = original_curr.next
print('None')


```

```python
curr= head
while curr and curr.next:
    if curr.val == curr.next.val:
        curr.next = curr.next.next
    else:
        curr = curr.next
print('Nodes after Removing the duplicates')
r_curr = head
while r_curr:
    print(r_curr.val)
    r_curr= r_curr.next
print('None')
    
```
