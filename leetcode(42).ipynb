42. Trapping Rain Water
<br>
Given n non-negative integers representing an elevation map where the width of each bar is 1, compute how much water it can trap after raining.

 
<br>
Example 1:
<br>

Input: height = [0,1,0,2,1,0,1,3,2,1,2,1]<br>
Output: 6<br>
Explanation: The above elevation map (black section) is represented by array [0,1,0,2,1,0,1,3,2,1,2,1]. In this case, 6 units of rain water (blue section) are being trapped.

complete

```python
#this is brute force approach 
height = [4,2,0,3,2,5]
h = len(height)
count = 0
for i in range(h):
    left = 0
    right = h-1
    max_right = float('-inf')
    max_left = float('-inf')
    while right>i:
        if height[right]>max_right:
            max_right = height[right]
        right-=1
    while left<i:
        if height[left]>max_left:
            max_left = height[left]
        left+=1
    total = min(max_right,max_left)-height[i]
    if total>=0:
        count+=total
count
    
        
```

```python
# this is prefix and suffix approach saves time complexity
nums = [0,1,0,2,1,0,1,3,2,1,2,1]
n = len(nums)
left = 0
right = n-1
maxs_left = [0]*n
maxs_right = [0]*n
m_left= float('-inf')
m_right= float('-inf')
while left<n and right>=0:
    if nums[left]>=m_left:
        maxs_left[left] = nums[left]
        m_left = nums[left]
        
    if nums[right]>=m_right:
        maxs_right[right] = nums[right]
        m_right = nums[right]
       
    maxs_left[left] = m_left
    maxs_right[right]= m_right
    left+=1
    right-=1
c = 0
for i in range(n):
    t = min(maxs_left[i],maxs_right[i])-nums[i]
    if t>0:
        c+=t
c
    
```
