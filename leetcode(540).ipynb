Single Element in a Sorted Array<br><br>You are given a sorted array consisting of only integers where every element appears exactly twice, except for one element which appears exactly once.<br><br>Return the single element that appears only once.<br><br>Your solution must run in O(log n) time and O(1) space.<br><br>Example 1:<br>Input<br>nums = [1,1,2,3,3,4,4,8,8]<br>Output<br>2<br><br>Example 2:<br>Input<br>nums = [3,3,7,7,10,11,11]<br>Output<br>10<br><br>Constraints:<br>- 1 <= nums.length <= 105<br>- 0 <= nums[i] <= 105

```python
nums = [1,1,2,3,3,4,4,8,8]
i = 0
j = len(nums) - 1
while i<j:
    mid = (i+j) //2
    if mid % 2 == 1: # if mid is odd make it even 
        mid -=1 # -=1 cause we dont have to worry about index out of range
    if nums[mid] == nums[mid+1]:
        i = mid + 2
    else:
        j = mid
nums[i]
```
