Given two strings s and t, return true if t is an anagram of s, and false otherwise.

print('complete')

```python
s = "anagram"
t = "nagaram"
n = len(s)
dict_s={}
dict_t ={}
for ch in s:
    if ch in dict_s:
        dict_s[ch]+=1
    else:
        dict_s[ch]=1

for ch in t:
    if ch in dict_t:
        dict_t[ch]+=1
    else:
        dict_t[ch]=1


for c in dict_s:
    if c not in dict_t or dict_s[c]!=dict_t[c]:
        print(False)
        break
else:
    print(True)
```
