1094. Car Pooling
<br>
There is a car with capacity empty seats. The vehicle only drives east (i.e., it cannot turn around and drive west).
<br>
You are given the integer capacity and an array trips where trips[i] = [numPassengersi, fromi, toi] indicates that the ith trip has numPassengersi passengers and the locations to pick them up and drop them off are fromi and toi respectively. The locations are given as the number of kilometers due east from the car's initial location.
<br>
Passengers are dropped off before new passengers are picked up at the same location. At every point along the route, the total number of passengers in the car must not exceed capacity.
<br>
Return true if it is possible to pick up and drop off all passengers for all the given trips, or false otherwise.

 <br><br>

Example 1:<br>

Input: trips = [[2,1,5],[3,3,7]], capacity = 4<br>
Output: false<br><br>
Explanation:<br>
At kilometer 1, 2 passengers are picked up, so the car holds 2.<br>
At kilometer 3, 3 more are picked up, so the car holds 5.<br>
Since 5 > capacity = 4, the trips cannot all be completed.<br>
Example 2:<br>

Input: trips = [[2,1,5],[3,3,7]], capacity = 5<br>
Output: true<br>
Explanation:<br>
At kilometer 1, the car holds 2 passengers.<br>
At kilometer 3, the car holds 5 passengers.<br>
At kilometer 5, the first 2 are dropped off, so the car holds 3.<br>
At kilometer 7, the last 3 are dropped off, so the car holds 0.<br>
The maximum occupancy is 5, which never exceeds capacity = 5.<br>
 

Constraints:<br>

1 <= trips.length <= 1000<br>
trips[i].length == 3<br>
1 <= numPassengersi <= 100<br>
0 <= fromi < toi <= 1000<br>
1 <= capacity <= 105<br>

```python
trips = [[2,1,5],[3,3,7]]
capacity = 4
```

```python
# why 1002: just fuckin see the constraints
pre= [0]*1002
max_end = 0
for trip in trips:
    passenger  = trip[0]
    start = trip[1]
    end = trip[2]
    pre[start]+=passenger
    pre[end]-=passenger
    if max_end<end:
        max_end = end

passenger = 0
for i in range(1002):
    passenger+=pre[i]
    if passenger>capacity:
        print(False)
        break
else:
    print(True)
```

```python
# fail what if all the trip overlap  
pas = trips[0][0]
n = len(trips)
for i in range(1,n):
    prev = trips[i-1]
    next = trips[i]
    if next[1]<prev[2]:
        pas+=next[0]
    else:
        pas = pas - prev[0]
        pas+=next[0]

if pas<=capacity:
    print(True)
else:
    print(False)
        
    
```
