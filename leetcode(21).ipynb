21. Merge Two Sorted Lists
<br>
You are given the heads of two sorted linked lists list1 and list2.
<br>
Merge the two lists into one sorted list. The list should be made by splicing together the nodes of the first two lists.
<br>
Return the head of the merged linked list.
<br>
 

Example 1:<br>

<br>
Input: list1 = [1,2,4], list2 = [1,3,4]<br>
Output: [1,1,2,3,4,4]<br>

```python
class Node:
    def __init__(self,val ,next = None):
        self.val = val
        self.next = next
        
```

```python
list1 = [1,2,4]
list2 = [1,3,4]

def fill_linked_list(values):
    n = len(values)
    head =None
    tail = None
    for i in range(n):
        new_node = Node(values[i])
        if head is None:
            head = new_node
            tail = new_node
        else:
            tail.next = new_node
            tail = new_node
    return head

head1 = fill_linked_list(list1)
head2 = fill_linked_list(list2)

```

```python
current = head1
while current!=None:
    print(current.val)
    current = current.next
```

```python
dummy = Node(0)


temp1 = head1
temp2 = head2
temp3 = dummy
while temp1!= None and temp2!= None:
    if temp1.val<temp2.val:
        temp3.next = temp1
        temp1 = temp1.next
    else:
        temp3.next = temp2
        temp2 = temp2.next
    temp3 = temp3.next
if temp1!=None:
    temp3.next = temp1
else:
    temp3.next = temp2
    
head = dummy.next

current = head

while current!=None:
    print(current.val)
    current = current.next
```
