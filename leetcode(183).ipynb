183. Customers Who Never Order

```python
import pandas as pd

# Customers table
customers = pd.DataFrame({
    'id': [1, 2, 3, 4],
    'name': ['Joe', 'Henry', 'Sam', 'Max']
})

# Orders table
orders = pd.DataFrame({
    'id': [1, 2],
    'customerId': [3, 1]
})
```

```python
merge = customers.merge(orders,left_on = 'id',right_on = 'customerId',how = 'left')
result = merge[merge['customerId'].isnull()][['name']].rename(columns={'name':'Customers'})
```
