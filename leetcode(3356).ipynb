3356. Zero Array Transformation II
<br>
You are given an integer array nums of length n and a 2D array queries where queries[i] = [li, ri, vali].
<br>
Each queries[i] represents the following action on nums:
<br>
Decrement the value at each index in the range [li, ri] in nums by at most vali.
The amount by which each value is decremented can be chosen independently for each index.
A Zero Array is an array with all its elements equal to 0.
<br>
Return the minimum possible non-negative value of k, such that after processing the first k queries in sequence, nums becomes a Zero Array. If no such k exists, return -1.
<br>
 
<br>
Example 1:
<br>
Input: nums = [2,0,2], queries = [[0,2,1],[0,2,1],[1,1,3]]
<br>
Output: 2
<br>
Explanation:
<br>
For i = 0 (l = 0, r = 2, val = 1):<br>
Decrement values at indices [0, 1, 2] by [1, 0, 1] respectively.<br>
The array will become [1, 0, 1].<br>
For i = 1 (l = 0, r = 2, val = 1):<br>
Decrement values at indices [0, 1, 2] by [1, 0, 1] respectively.<br>
The array will become [0, 0, 0], which is a Zero Array. Therefore, the minimum value of k is 2.
Example 2:<br>

Input: nums = [4,3,2,1], queries = [[1,3,2],[0,2,1]]
<br>
Output: -1<br>

Explanation:<br>
For i = 0 (l = 1, r = 3, val = 2):<br>
Decrement values at indices [1, 2, 3] by [2, 2, 1] respectively.<br>
The array will become [4, 1, 0, 0].<br>
For i = 1 (l = 0, r = 2, val = 1):<br>
Decrement values at indices [0, 1, 2] by [1, 1, 0] respectively.<br>
The array will become [3, 0, 0, 0], which is not a Zero Array.<br>
 

Constraints:<br>
<br>
1 <= nums.length <= 105<br>
0 <= nums[i] <= 5 * 105<br>
1 <= queries.length <= 105<br>
queries[i].length == 3<br>
0 <= li <= ri < nums.length<br>
1 <= vali <= 5<br>

```python
nums = [7,6,8]
queries =[[0,0,2],[0,1,5],[2,2,5],[0,2,4]]
```

```python
def new_checker(k,nums,queries):
    n = len(nums)
    diff = [0]*(n+1)
    
    for i in range(k):
        l,r,val = queries[i]
        diff[l]+=val
        diff[r+1]-=val
    cover = 0
    for i in range(n):
        cover+=diff[i]
        if nums[i]>cover:
            return False
    return True

i = 0
j = len(queries) 
while i<j:
    mid = (i+j)//2
    if new_checker(mid,nums,queries):
        print(mid)
        j = mid
    else:
        i = mid +1


if new_checker(i,nums,queries):
    print(i)
else:
    print(-1)

```
