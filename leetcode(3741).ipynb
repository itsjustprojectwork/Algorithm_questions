```python
nums = [5,5,5,2,5]
n = len(nums)
```

```python
# this aint happing cause if there are more than 3 same numbers this algorithm fails
n = len(nums)
m_d = float('inf')
for i in range(n-2):
    j = i+1
    k = n-1
    while (j<k):
        if nums[i]!=nums[j]:
            j+=1
        if nums[i]!=nums[k]:
            k-=1
        if nums[i]==nums[j]==nums[k] and k!=j:
            x = abs(i - j) + abs(j - k) + abs(k - i)
            if x<m_d:
                m_d = x
            j+=1
            k-=1
                
m_d
```

 abs(i - j) + abs(j - k) + abs(k - i)<br><br>
 so the idea is<br><br> |i-j| + |j-k| + |k-i| == |j-i| + |k-j| + |k-i|<br> --> j - i + k - j + k - i<br> --> 2(k-i)<br>

```python
# this is O(n) approach 
seen_indexes = {}
for i in range(n):
    if nums[i] in seen_indexes:
        seen_indexes[nums[i]].append(i)
    else:
        seen_indexes[nums[i]] = [i]
        
m = float('inf')
# this anit O(n^2) cause inner loop goes for n-2 times 
for index in seen_indexes:
    x =seen_indexes[index]
    for i in range(len(x)-2):
        d = 2 * abs(x[i+2]-x[i])
        if d<m:
            m = d 
m
            
```
