
8. String to Integer (atoi)
<br>
Implement the myAtoi(string s) function, which converts a string to a 32-bit signed integer.
<br>
The algorithm for myAtoi(string s) is as follows:
<br>
Whitespace: Ignore any leading whitespace (" ").<br>
Signedness: Determine the sign by checking if the next character is '-' or '+', assuming positivity if neither present.<br>
Conversion: Read the integer by skipping leading zeros until a non-digit character is encountered or the end of the string is reached. If no digits were read, then the result is 0.<br>
Rounding: If the integer is out of the 32-bit signed integer range [-231, 231 - 1], then round the integer to remain in the range. Specifically, integers less than -231 should be rounded to -231, and integers greater than 231 - 1 should be rounded to 231 - 1.
Return the integer as the final result.

complete

```python
import string
class Solution:
    def myAtoi(self, s: str) -> int:
        s = s.strip()
        if len(s)<1:
            return 0

        sign = -1 if s[0]=='-' else 1
        if s[0]=='-' or s[0]=='+':
            s = s[1::]

        num = 0

        for i in range(len(s)):
            if s[i] not in string.digits:
                break
            num = num *10 + int(s[i])
        
        num = sign*num
        if num > 2**31 -1:
            return 2**31 -1
        elif num < -2**31:
            return -2**31
        return num
```
