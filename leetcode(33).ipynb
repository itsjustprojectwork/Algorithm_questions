33. Search in Rotated Sorted Array<br><br>There is an integer array nums sorted in ascending order (with distinct values).<br><br>Prior to being passed to your function, nums is possibly left rotated at an unknown index k (1 <= k < nums.length) such that the resulting array is [nums[k], nums[k+1], ..., nums[n-1], nums[0], nums[1], ..., nums[k-1]] (0-indexed). For example, [0,1,2,4,5,6,7] might be left rotated by 3 indices and become [4,5,6,7,0,1,2].<br><br>Given the array nums after the possible rotation and an integer target, return the index of target if it is in nums, or -1 if it is not in nums.<br><br>You must write an algorithm with O(log n) runtime complexity.<br><br>Example 1:<br>Input<br>nums = [4,5,6,7,0,1,2], target = 0<br>Output<br>4<br><br>Example 2:<br>Input<br>nums = [4,5,6,7,0,1,2], target = 3<br>Output<br>-1<br><br>Example 3:<br>Input<br>nums = [1], target = 0<br>Output<br>-1<br><br>Constraints:<br>- 1 <= nums.length <= 5000<br>- -104 <= nums[i] <= 104<br>- All values of nums are unique.<br>- nums is an ascending array that is possibly rotated.<br>- -104 <= target <= 104

```python
# there are two approaches to solve this dumb question and both get accepted

# Approach 1 O(n)
nums = [4,5,6,7,0,1,2]
target = 0

n = len(nums)
for i in range(n):
    if nums[i]==target:
        print(i)
        break
else:
    print(-1)
```

```python
# approach 2 O(n//2)
nums = [4,5,6,7,0,1,2]
target = 0

i = 0
j = len(nums)-1
while i<=j:
    mid = (i+j) //2 
    if nums[mid] == target:
        print(mid)
        break
    if nums[i]<nums[mid]:
        if nums[i]<=target<nums[mid]:
            j = mid -1
        else:
            i = mid +1
    else:
        if nums[mid]<target<=nums[j]:
            i = mid +1
        else:
            j = mid -1 
else:
    print(-1)
```
