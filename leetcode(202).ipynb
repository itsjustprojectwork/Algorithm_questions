202. Happy Number

```python
n = 19
def square_num(n):
    total = 0 
    while n>0:
        digit = n % 10
        total += digit*digit
        n//=10
    return total


seen = set()
while n!=1:
    if n in seen:
        print(False)
        break
    seen.add(n)
    n= square_num(n)
    
else:
    print(True)
        
```
