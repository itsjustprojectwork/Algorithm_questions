```python
rowindex = 3
n = rowindex +1
result =[]
for i in range(n):
    if i == 0:
        result.append([1])
        continue
    
    previous = result[-1]
    
    row = [1] # each row starts with 1
    for j in range(1,len(previous)):
        x = previous[j-1] + previous[j]
        row.append(x)
    row.append(1) # each row ends with 1
    
    result.append(row)
    
    
```
