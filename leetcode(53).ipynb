Maximum Subarray<br><br>Given an integer array nums, find the subarray with the largest sum, and return its sum.<br><br>Example 1:<br>Input<br>nums = [-2,1,-3,4,-1,2,1,-5,4]<br>Output<br>6<br>Explanation<br>The subarray [4,-1,2,1] has the largest sum 6.<br><br>Example 2:<br>Input<br>nums = [1]<br>Output<br>1<br>Explanation<br>The subarray [1] has the largest sum 1.<br><br>Example 3:<br>Input<br>nums = [5,4,-1,7,8]<br>Output<br>23<br>Explanation<br>The subarray [5,4,-1,7,8] has the largest sum 23.<br><br>Constraints:<br>- 1 <= nums.length <= 105<br>- -104 <= nums[i] <= 104<br>- Follow up: If you have figured out the O(n) solution, try coding another solution using the divide and conquer approach, which is more subtle.

```python
nums = [-2,1,-3,4,-1,2,1,-5,4]
curren_sum = 0
max_sub_sum = float('-inf')
for num in nums:
    curren_sum+=num
    if curren_sum>max_sub_sum:
        max_sub_sum = curren_sum
    if curren_sum<0:
        curren_sum = 0
max_sub_sum
```
