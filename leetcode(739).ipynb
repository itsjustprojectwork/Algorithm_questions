739. Daily Temperatures<br><br>Given an array of integers temperatures represents the daily temperatures, return an array answer such that answer[i] is the number of days you have to wait after the ith day to get a warmer temperature. If there is no future day for which this is possible, keep answer[i] == 0 instead.<br><br>Example 1:<br>Input<br>temperatures = [73,74,75,71,69,72,76,73]<br>Output<br>[1,1,4,2,1,1,0,0]<br><br>Example 2:<br>Input<br>temperatures = [30,40,50,60]<br>Output<br>[1,1,1,0]<br><br>Example 3:<br>Input<br>temperatures = [30,60,90]<br>Output<br>[1,1,0]<br><br>Constraints:<br>- 1 <= temperatures.length <= 105<br>- 30 <= temperatures[i] <= 100

```python
temperatures = [73,74,75,71,69,72,76,73]
```

```python
# brute force approach

n = len(temperatures)
ans = []
for i in range(n):
    for j in range(i+1,n):
        if temperatures[i]<temperatures[j]:
            ans.append(j-i)
            break
    else:
        ans.append(0)

ans
    
```

```python
stack = []
n = len(temperatures)
ans = [0] * n
for i in range(n-1,-1,-1):
    while stack:
        if temperatures[i]>= temperatures[stack[-1]]:
            stack.pop()
        else:
            distance = stack[-1] - i
            ans[i] = distance
            break
    stack.append(i)
ans
```
