744. Find Smallest Letter Greater Than Target<br><br>You are given an array of characters letters that is sorted in non-decreasing order, and a character target. There are at least two different characters in letters.<br><br>Return the smallest character in letters that is lexicographically greater than target. If such a character does not exist, return the first character in letters.<br><br>Example 1:<br>Input<br>letters = ["c","f","j"], target = "a"<br>Output<br>"c"<br>Explanation<br>The smallest character that is lexicographically greater than 'a' in letters is 'c'.<br><br>Example 2:<br>Input<br>letters = ["c","f","j"], target = "c"<br>Output<br>"f"<br>Explanation<br>The smallest character that is lexicographically greater than 'c' in letters is 'f'.<br><br>Example 3:<br>Input<br>letters = ["x","x","y","y"], target = "z"<br>Output<br>"x"<br>Explanation<br>There are no characters in letters that is lexicographically greater than 'z' so we return letters[0].<br><br>Constraints:<br>- 2 <= letters.length <= 104<br>- letters[i] is a lowercase English letter.<br>- letters is sorted in non-decreasing order.<br>- letters contains at least two different characters.<br>- target is a lowercase English letter.

```python
letters = ["c","f","j"]
target = "c"
i = 0
j = len(letters) -1

while i<j:
    mid = (i+j)//2
    char = letters[mid]
    if ord(char)<= ord(target):
        i = mid +1
    else:
        j = mid

if ord(letters[j])>ord(target):
    print(letters[j])
else:
    print(letters[0])
```
