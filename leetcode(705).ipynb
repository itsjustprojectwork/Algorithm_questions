```python
class MyHashSet:
    def __init__(self):
        self.nums = []
    
    def add(self, key: int) -> None:
        l = len(self.nums)
        if key not in self.nums:
            new_nums = [0]*(l+1)
            for i in range(l):
                new_nums[i]= self.nums[i]
            new_nums[-1] = key
            self.nums = new_nums
            
    def remove(self, key: int) -> None:
        l = len(self.nums)
        if key in self.nums:
            new_nums = [0]*(l-1)
            j =0
            i =0
            while j<l-1:
                if self.nums[i] == key:
                    i+=1
                else:
                    new_nums[j] = self.nums[i]
                    i+=1
                    j+=1
            self.nums = new_nums
        
            
    def contains(self, key: int) -> bool:
        for num in self.nums:
            if num == key:
                return True
        return False
    
    def print(self):
        print(self.nums)
```

```python
for i in range(len(p)):
    if p[i]=='MyHashSet':
        has = MyHashSet()
        print(i)
    elif p[i]=='add':
        val = v[i][0]
        has.add(val)
        print(i)
    elif p[i]=='contains':
        val = v[i][0]
        print(i,has.contains(val))
        
    elif p[i] == 'remove':
        val = v[i][0]
        has.remove(val)
        print(i)
```
