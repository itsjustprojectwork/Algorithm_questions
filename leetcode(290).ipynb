290. Word Pattern
<br>
Given a pattern and a string s, find if s follows the same pattern.
<br>
Here follow means a full match, such that there is a bijection between a letter in pattern and a non-empty word in s. Specifically:
<br>
Each letter in pattern maps to exactly one unique word in s.<br>
Each unique word in s maps to exactly one letter in pattern.<br>
No two letters map to the same word, and no two words map to the same letter.<br>

complete

```python
p = "abac"
s = "dog cat cat dog"
s = list(s.strip().split())
 
```

19/3

```python
d_sp={}
d_ps={}
n = len(p)
m = len(s)

for i in range(n):
    if s[i] in d_sp:
        if d_sp[s[i]]!=p[i]:
            print(False)
            break
    if p[i] in d_ps:
        if d_ps[p[i]]!=s[i]:
            print(False)
            break
    d_sp[s[i]] = p[i]
    d_ps[p[i]] = s[i]
else:
    print(True)

    
```
