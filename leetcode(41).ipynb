41. First Missing Positive<br><br>Given an unsorted integer array nums. Return the smallest positive integer that is not present in nums.<br><br>You must implement an algorithm that runs in O(n) time and uses O(1) auxiliary space.<br><br>Example 1:<br>Input<br>nums = [1,2,0]<br>Output<br>3<br>Explanation<br>The numbers in the range [1,2] are all in the array.<br><br>Example 2:<br>Input<br>nums = [3,4,-1,1]<br>Output<br>2<br>Explanation<br>1 is in the array but 2 is missing.<br><br>Example 3:<br>Input<br>nums = [7,8,9,11,12]<br>Output<br>1<br>Explanation<br>The smallest positive integer 1 is missing.<br><br>Constraints:<br>- 1 <= nums.length <= 105<br>- -231 <= nums[i] <= 231 - 1

```python
nums = [1]
n = len(nums)
freq = {}
for num in nums:
    if num in freq:
        freq[num]+=1
    else:
        freq[num] = 1
        
for i in range(1,n+2):
    if i in freq:
        continue
    else:
        print(i)
        break
```
