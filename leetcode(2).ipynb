2. Add Two Numbers
<br>
You are given two non-empty linked lists representing two non-negative integers. The digits are stored in reverse order, and each of their nodes contains a single digit. Add the two numbers and return the sum as a linked list.
<br>
You may assume the two numbers do not contain any leading zero, except the number 0 itself.
<br>
 
<br>
Example 1:


Input: l1 = [2,4,3], l2 = [5,6,4]<br>
Output: [7,0,8]<br>
Explanation: 342 + 465 = 807.<br>
Example 2:
<br>
Input: l1 = [0], l2 = [0]<br>
Output: [0]<br>
Example 3:<br>

Input: l1 = [9,9,9,9,9,9,9], l2 = [9,9,9,9]<br>
Output: [8,9,9,9,0,0,0,1]<br>


```python
class Node:
    def __init__(self,val,next= None):
        self.val = val
        self.next = next
    
def inset_val_list(values):
    head = None
    tail = None
    for num in values:
        node = Node(num)
        if head is None:
            head = node
            tail = node
        else:
            tail.next = node
            tail= node
    return head
    
def reverse(l1):
    prev = None
    curr = l1
    while curr!=None:
        temp = curr.next
        curr.next = prev
        prev = curr
        curr = temp
    return prev

def count_nodes(head):
    curr = head
    count = 0
    while curr!=None:
        count = count*10 + curr.val
        curr = curr.next
    return count

def print_nodes(head):
    curr = head
    count = 0
    while curr!=None:
        print(curr.val)
        curr = curr.next
        
def inser_val(n):
    val = []
    if n==0:
        val.append(0)
        return val
    while n!=0:
        d = n%10
        val.append(d)
        n//=10
    return val
```

```python
v1 = [0]
v2 = [0]
l1 = inset_val_list(v1)
l2 = inset_val_list(v2)


```

```python
rev_l1 = reverse(l1)
rev_l2  = reverse(l2)

```

```python
print(print_nodes(rev_l1))
print()
print(print_nodes(rev_l2))
```
