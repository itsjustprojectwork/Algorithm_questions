<br><br>
Given an integer array nums, find three numbers whose product is maximum and return the maximum product.

 <br><br>

Example 1:<br>
<br>
Input: nums = [1,2,3]<br>
Output: 6<br>

```python
# TLE 
nums=[2,7,6,5,8,8,-2,-4,-6,-8]
n = len(nums)
for _ in range(n):
    for i in range(n):
        for j in range(n-i-1):
            if nums[j]>nums[j+1]:
                nums[j+1],nums[j] = nums[j],nums[j+1]

x = nums[0]*nums[1]*nums[n-1]
m = nums[n-1]*nums[n-2]*nums[n-3]
if x>m:
    print(x)
else:
    print(m)
```
