K-Concatenation Maximum Sum<br><br>Given an integer array arr and an integer k, modify the array by repeating it k times.<br><br>For example, if arr = [1, 2] and k = 3 then the modified array will be [1, 2, 1, 2, 1, 2].<br><br>Return the maximum sub-array sum in the modified array. Note that the length of the sub-array can be 0 and its sum in that case is 0.<br><br>As the answer can be very large, return the answer modulo 10**9 + 7.<br><br>Example 1:<br>Input<br>arr = [1,2], k = 3<br>Output<br>9<br><br>Example 2:<br>Input<br>arr = [1,-2,1], k = 5<br>Output<br>2<br><br>Example 3:<br>Input<br>arr = [-1,-2], k = 7<br>Output<br>0<br><br>Constraints:<br>- 1 <= arr.length <= 105<br>- 1 <= k <= 105<br>- -104 <= arr[i] <= 104

```python
arr =[2,5]
k = 2
```

```python
def kadanes_algo(nums):
    current = 0
    max_sub = float('-inf')
    for num in nums:
        current +=num
        if current<0:
            current = 0
        if current > max_sub:
            max_sub = current
    return max_sub

if k==1:
    print(kadanes_algo(arr))

double_max = kadanes_algo(arr*2)

# total_sum
total = 0
for num in arr:
    total+=num

MOD = 10**9 + 7

if total<=0:
    print(double_max % MOD)
else:
    print(double_max + (k-2)*total % MOD)
```

```python
# tle
curr_sum = 0
max_sub_sum = float('-inf')
for _ in range(k):
    for num in arr:
        curr_sum+=num
        if curr_sum<0:
            curr_sum = 0
        if curr_sum>max_sub_sum:
            max_sub_sum = curr_sum


MOD = 10**9 + 7

max_sub_sum % MOD
```
