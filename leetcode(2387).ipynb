2381. Shifting Letters II
<br>
You are given a string s of lowercase English letters and a 2D integer array shifts where shifts[i] = [starti, endi, directioni]. For every i, shift the characters in s from the index starti to the index endi (inclusive) forward if directioni = 1, or shift the characters backward if directioni = 0.
<br>
Shifting a character forward means replacing it with the next letter in the alphabet (wrapping around so that 'z' becomes 'a'). Similarly, shifting a character backward means replacing it with the previous letter in the alphabet (wrapping around so that 'a' becomes 'z').<br>

Return the final string after all such shifts to s are applied.
<br>
 

Example 1:<br>

Input: s = "abc", shifts = [[0,1,0],[1,2,1],[0,2,1]]<br>
Output: "ace"<br>
Explanation: Firstly, shift the characters from index 0 to index 1 backward. Now s = "zac".<br>
Secondly, shift the characters from index 1 to index 2 forward. Now s = "zbd".<br>
Finally, shift the characters from index 0 to index 2 forward. Now s = "ace".<br>
Example 2:<br>

Input: s = "dztz", shifts = [[0,0,0],[1,1,1]]<br>
Output: "catz"<br>
Explanation: Firstly, shift the characters from index 0 to index 0 backward. Now s = "cztz".<br>
Finally, shift the characters from index 1 to index 1 forward. Now s = "catz".<br>
 

Constraints:<br>
<br>
1 <= s.length, shifts.length <= 5 * 104<br>
shifts[i].length == 3<br>
0 <= starti <= endi < s.length<br>
0 <= directioni <= 1<br>
s consists of lowercase English letters.<br>

```python
S = "abc"
shifts = [[0,1,0],[1,2,1],[0,2,1]]
```

```python
n = len(S)
freq = {0: 'a',
 1: 'b',
 2: 'c',
 3: 'd',
 4: 'e',
 5: 'f',
 6: 'g',
 7: 'h',
 8: 'i',
 9: 'j',
 10: 'k',
 11: 'l',
 12: 'm',
 13: 'n',
 14: 'o',
 15: 'p',
 16: 'q',
 17: 'r',
 18: 's',
 19: 't',
 20: 'u',
 21: 'v',
 22: 'w',
 23: 'x',
 24: 'y',
 25: 'z'}
ans = [0]*(n+1)
for shift in shifts:
    start = shift[0]
    end = shift[1]
    direction = shift[2]
    if direction == 1:
        ans[start]+=1
        ans[end+1]-=1
    else:
        ans[start]-=1
        ans[end+1]+=1

pre = [0]*(n)
pre[0] = ans[0]
for i in range(1,n):
    pre[i] = pre[i-1] + ans[i]
pre
```
