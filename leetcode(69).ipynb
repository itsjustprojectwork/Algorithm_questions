69. Sqrt(x)<br><br>Given a non-negative integer x, return the square root of x rounded down to the nearest integer. The returned integer should be non-negative as well.<br><br>You must not use any built-in exponent function or operator.<br><br>- For example, do not use pow(x, 0.5) in c++ or x ** 0.5 in python.<br><br>Example 1:<br>Input<br>x = 4<br>Output<br>2<br>Explanation<br>The square root of 4 is 2, so we return 2.<br><br>Example 2:<br>Input<br>x = 8<br>Output<br>2<br>Explanation<br>The square root of 8 is 2.82842..., and since we round it down to the nearest integer, 2 is returned.<br><br>Constraints:<br>- 0 <= x <= 231 - 1

```python
x = 10

if x==0:
    print(0)
elif x<=2:
    print(1)
else:
    i= 1
    j = x
    while i<=j:
        mid= (i+j) //2
        if mid * mid <=x:
            val = mid
            i = mid +1
        else:
            j = mid -1
    print(val)
```
