19. Remove Nth Node From End of List
<br>
Given the head of a linked list, remove the nth node from the end of the list and return its head.
<br>
 
<br>
Example 1:
<br>

Input: head = [1,2,3,4,5], n = 2<br>
Output: [1,2,3,5]<br>
Example 2:<br>
<br>
Input: head = [1], n = 1<br>
Output: []<br>
Example 3:<br>
<br>
Input: head = [1,2], n = 1<br>
Output: [1]<br>

```python
class Node:
    def __init__(self,val = 0,next =None):
        self.val = val
        self.next = next
        
        
def print_all_node(head):
    curr = head
    while curr!=None:
        print(curr.val)

        curr = curr.next
        
    
def length_of_head(head):
    curr = head
    l = 0
    while curr!=None:
        l+=1
        curr = curr.next
    return l
```

```python
# store the nodes

values = [2,3,4,5,6,7,8]
n = len(values)
head = None
tail = None
for i in range(n):
    node = Node(values[i])
    if head is None:
        head = node
        tail = node
    else:
        tail.next = node
        tail = node
```

```python
print_all_node(head)
```

```python
# approach one to solve the problem using brute force
l = length_of_head(head)
k = 2
e = l-k-1
c  = head
for i in range(e):
    c = c.next
    
c.next = c.next.next

c = head
print_all_node(c)
```

```python
# approach 2 using fast-slow method
fast = head
slow = head
k = 2 # we will remove 2nd last elemtnt taht is 7 and we need to control 6 

# here will move fast by k times in order that our slow reach the previous elemnt we want to remove

for i in range(k+1): # we moved our fast pointer k times now our pointer is on 4
    fast = fast.next
    
# we move both the pointes together one by one
while fast:
    slow = slow.next
    fast = fast.next
    
print(slow.val) # see our slow pointer is on 6 and we need to remove next element

slow.next = slow.next.next
# done 
print_all_node(head)
```

Down problem is same so edge case in order to avoid error we use dummy


```python
# Definition for singly-linked list.
# class ListNode:
#     def __init__(self, val=0, next=None):
#         self.val = val
#         self.next = next
class Solution:
    def removeNthFromEnd(self, head: Optional[ListNode], n: int) -> Optional[ListNode]:
        dummy = ListNode(0)
        dummy.next = head
        # approach 2 using fast-slow method
        fast = dummy
        slow = dummy

        # here will move fast by k times in order that our slow reach the previous elemnt we want to remove

        for i in range(n+1): # we moved our fast pointer k times now our pointer is on 4
            fast = fast.next
            
        # we move both the pointes together one by one
        while fast:
            slow = slow.next
            fast = fast.next
            
        print(slow.val) # see our slow pointer is on 6 and we need to remove next element

        slow.next = slow.next.next
        # done 
        return dummy.next
```
