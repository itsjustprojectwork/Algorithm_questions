57. Insert Interval
<br>
You are given an array of non-overlapping intervals intervals where intervals[i] = [starti, endi] represent the start and the end of the ith interval and intervals is sorted in ascending order by starti. You are also given an interval newInterval = [start, end] that represents the start and end of another interval.
<br>
Two intervals are considered overlapping if they share at least one point.
<br>
Insert newInterval into intervals such that intervals is still sorted in ascending order by starti and intervals still does not have any overlapping intervals (merge overlapping intervals if necessary).
<br>
Return intervals after the insertion.
<br>
Note that you don't need to modify intervals in-place. You can make a new array and return it.
<br>
 <br>
<br>
Example 1:
<br>
Input: intervals = [[1,3],[6,9]], newInterval = [2,5]<br>
Output: [[1,5],[6,9]]<br>
Example 2:<br>
<br>
Input: intervals = [[1,2],[3,5],[6,7],[8,10],[12,16]], newInterval = [4,8]<br>
Output: [[1,2],[3,10],[12,16]]<br>
Explanation: Because the new interval [4,8] overlaps with [3,5],[6,7],[8,10].<br>
 
<br>
Constraints:<br>

0 <= intervals.length <= 104<br><br>
intervals[i].length == 2<br>
0 <= starti <= endi <= 105<br>
intervals is sorted by starti in ascending order.<br>
newInterval.length == 2<br>
0 <= start <= end <= 105<br>

```python
intervals = [[1,2],[3,5],[6,7],[8,10],[12,16]]
newInterval = [4,8]
result = []
i = 0
n = len(intervals)
# stop where inreval and new interval overlaps
while i<n and intervals[i][1]<newInterval[0]:
    result.append(intervals[i])
    i+=1

# handling ovelapping intevals
while i<n and intervals[i][0]<=newInterval[1]:
    newInterval[0] = min(intervals[i][0],newInterval[0])
    newInterval[1] = max(intervals[i][1],newInterval[1])
    i+=1
result.append(newInterval)

# appending ramaning intervals
while i<n:
    result.append(intervals[i])
    i+=1

result
    
```
