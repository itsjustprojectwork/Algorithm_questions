
5. Longest Palindromic Substring
<br>
Hint
Given a string s, return the longest palindromic substring in s.
<br>
 

Example 1:<br>

Input: s = "babad"<br>
Output: "bab"<br>
Explanation: "aba" is also a valid answer.<br>
Example 2:<br>
<br>
Input: s = "cbbd"<br>
Output: "bb"<br>

```python
def s_palin(s):
    return s==s[::-1]
```

```python
# this is O(n^3) approrach 
s = 'abbc'
n = len(s)
max_n = float('-inf')
s_b =''
for i in range(n):
    j=i+1
    while j<=n:
        k = len(s[i:j])
        if k>1:
            if s_palin(s[i:j]):
                if k>max_n:
                    max_n = k
                    s_b = s[i:j]
        j+=1

print(s_b)
    
```
