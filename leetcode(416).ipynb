
416. Partition Equal Subset Sum
<br>
Given an integer array nums, return true if you can partition the array into two subsets such that the sum of the elements in both subsets is equal or false otherwise.
<br>
 
<br>
Example 1:
<br>
Input: nums = [1,5,11,5]
Output: true<br>
Explanation: The array can be partitioned as [1, 5, 5] and [11].<br>
Example 2:<br>

Input: nums = [1,2,3,5]<br>
Output: false<br>
Explanation: The array cannot be partitioned into equal sum subsets.<br>
 <br>

Constraints:<br>
<br>
1 <= nums.length <= 200<br>
1 <= nums[i] <= 100<br>

```python
nums = [1,2,3,8]
```

```python
s = sum(nums)
if s%2!=0:
    print(False)
else:
    target = s//2
    dp = [False]*(target+1)
    dp[0] = True
    for num in nums:
        for j in range(target,num-1,-1):
            if dp[j-num]:
                dp[j] = True
    print(dp[target])
```

```python
# fail we are only checking pair which hit the target
s = sum(nums)
freq = {}
for num in nums:
    needed = s - num
    if needed in freq:
        print(True)
        break
    else:
        freq[needed] = 1
else:
    print(False)
```
