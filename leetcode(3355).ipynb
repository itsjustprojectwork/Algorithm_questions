3355. Zero Array Transformation I
<br>
You are given an integer array nums of length n and a 2D array queries, where queries[i] = [li, ri].
<br>
For each queries[i]:
<br>
Select a subset of indices within the range [li, ri] in nums.<br>
Decrement the values at the selected indices by 1.<br>
A Zero Array is an array where all elements are equal to 0.<br>
<br>
Return true if it is possible to transform nums into a Zero Array after processing all the queries sequentially, otherwise return false.
<br>
 
<br>
Example 1:<br>
<br>
Input: nums = [1,0,1], queries = [[0,2]]
<br>
Output: true
<br>
Explanation:
<br>
For i = 0:<br>
Select the subset of indices as [0, 2] and decrement the values at these indices by 1.<br>
The array will become [0, 0, 0], which is a Zero Array.<br>
Example 2:<br>
<br>
Input: nums = [4,3,2,1], queries = [[1,3],[0,2]]<br>

Output: false<br>

Explanation:<br>

For i = 0:<br>
Select the subset of indices as [1, 2, 3] and decrement the values at these indices by 1.<br>
The array will become [4, 2, 1, 0].<br>
For i = 1:<br>
Select the subset of indices as [0, 1, 2] and decrement the values at these indices by 1.<br>
The array will become [3, 1, 0, 0], which is not a Zero Array.<br>
 

Constraints:<br>

1 <= nums.length <= 105<br>
0 <= nums[i] <= 105<br>
1 <= queries.length <= 105<br>
queries[i].length == 2<br>
0 <= li <= ri < nums.length<br>

```python
nums = [1,0,1]
queries = [[0,2]]
```

```python
n = len(nums)
coverage = [0]*(n+1)
for query in queries:
    l = query[0]
    r = query[1]
    coverage[l]+=1
    coverage[r+1]-=1
pre = [0]*n
pre[0]= coverage[0]
for i in range(1,n):
    pre[i] = pre[i-1] + coverage[i]

for i in range(n):
    if nums[i]>pre[i]:
        print(False)
        break
else:
    print(True)
    
```
