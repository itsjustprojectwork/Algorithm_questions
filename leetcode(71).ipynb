71. Simplify Path<br><br>You are given an absolute path for a Unix-style file system, which always begins with a slash '/'. Your task is to transform this absolute path into its simplified canonical path.<br><br>The rules of a Unix-style file system are as follows:<br><br>- A single period '.' represents the current directory.<br>- A double period '..' represents the previous/parent directory.<br>- Multiple consecutive slashes such as '//' and '///' are treated as a single slash '/'.<br>- Any sequence of periods that does not match the rules above should be treated as a valid directory or file name. For example, '...' and '....' are valid directory or file names.<br><br>The simplified canonical path should follow these rules:<br><br>- The path must start with a single slash '/'.<br>- Directories within the path must be separated by exactly one slash '/'.<br>- The path must not end with a slash '/', unless it is the root directory.<br>- The path must not have any single or double periods ('.' and '..') used to denote current or parent directories.<br><br>Return the simplified canonical path.<br><br>Example 1:<br>Input<br>path = "/home/"<br>Output<br>"/home"<br>Explanation<br>The trailing slash should be removed.<br><br>Example 2:<br>Input<br>path = "/home//foo/"<br>Output<br>"/home/foo"<br>Explanation<br>Multiple consecutive slashes are replaced by a single one.<br><br>Example 3:<br>Input<br>path = "/home/user/Documents/../Pictures"<br>Output<br>"/home/user/Pictures"<br>Explanation<br>A double period ".." refers to the directory up a level (the parent directory).<br><br>Example 4:<br>Input<br>path = "/../"<br>Output<br>"/"<br>Explanation<br>Going one level up from the root directory is not possible.<br><br>Example 5:<br>Input<br>path = "/.../a/../b/c/../d/./"<br>Output<br>"/.../b/d"<br>Explanation<br>"..." is a valid name for a directory in this problem.<br><br>Constraints:<br>- 1 <= path.length <= 3000<br>- path consists of English letters, digits, period '.', slash '/' or '_'.<br>- path is a valid absolute Unix path.

```python
path ="/../a/c"
path=path.split("/")

stack = []

for s in path:
    if s=='' or s =='.':
        continue
    elif s == '..':
        if len(stack)>0:
            stack.pop()
    else:
        stack.append(s)

stack.insert(0,'')
if len(stack)<=1:
    print('/')
else:
    print('/'.join(stack))
```
