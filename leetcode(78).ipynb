78. Subsets
<br>
Given an integer array nums of unique elements, return all possible subsets (the power set).
<br>
The solution set must not contain duplicate subsets. Return the solution in any order.
<br>
 

Example 1:
<br>
Input: nums = [1,2,3]<br>
Output: [[],[1],[2],[1,2],[3],[1,3],[2,3],[1,2,3]]<br>
Example 2:<br>
<br>
Input: nums = [0]<br>
Output: [[],[0]]

complete

```python
class Solution:
    def subsets(self, nums: List[int]) -> List[List[int]]:
        subset = [[]]
        for num in nums:
            new = []
            for s in subset:
                new.append(s+[num])
            subset+=new
        return subset
```

```python
# we got extra []
nums = [1,2,3]
subsets = [[]]
for n in nums:
    new_set = []
    for sub in subsets:
        new_set.append(sub+[n])
    subsets.append(new_set)
subsets
```
