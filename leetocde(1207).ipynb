
1207. Unique Number of Occurrences
<br>
Given an array of integers arr, return true if the number of occurrences of each value in the array is unique or false otherwise.
<br>
 
<br>
Example 1:
<br>
Input: arr = [1,2,2,1,1,3]<br>
Output: true<br>
Explanation: The value 1 has 3 occurrences, 2 has 2 and 3 has 1. No two values have the same number of occurrences.<br>
Example 2:<br>
<br>
Input: arr = [1,2]<br>
Output: false<br>
Example 3:<br>
<br>
Input: arr = [-3,0,1,-3,1,1,1,-3,10,0]<br>
Output: true<br>
 <br>

Constraints:<br>

1 <= arr.length <= 1000<br>
-1000 <= arr[i] <= 1000<br>

```python
arr = [1,2,2,1,3,1]
```

```python
freq = {}
for num in arr:
    if num in freq:
        freq[num]+=1
    else:
        freq[num]=1
    
freq
```

```python
seen = set()
for num in freq:
    if freq[num] in seen:
        print(False)
        break
    else:
        seen.add(freq[num])
else:
    print(True)
```
