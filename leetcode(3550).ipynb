3550. Smallest Index With Digit Sum Equal to Index
<br>
You are given an integer array nums.
<br>
Return the smallest index i such that the sum of the digits of nums[i] is equal to i.
<br>
If no such index exists, return -1.
<br>
 

Example 1:
<br>
Input: nums = [1,3,2]
<br>
Output: 2
<br>
Explanation:
<br>
For nums[2] = 2, the sum of digits is 2, which is equal to index i = 2. Thus, the output is 2.
Example 2:<br>
<br>
Input: nums = [1,10,11]<br>
<br>
Output: 1<br>
<br>
Explanation:<br>
<br>
For nums[1] = 10, the sum of digits is 1 + 0 = 1, which is equal to index i = 1.<br>
For nums[2] = 11, the sum of digits is 1 + 1 = 2, which is equal to index i = 2.<br>
Since index 1 is the smallest, the output is 1.<br>
Example 3:<br>
<br>
Input: nums = [1,2,3]<br><br>

Output: -1<br>

Explanation:<br>

Since no index satisfies the condition, the output is -1.<br>

```python
nums = [804,368,101,3]
n = len(nums)

def sum_digit(n):
    sum_d  = 0
    while n!=0:
        d = n%10
        sum_d += d
        n//=10
        
    return sum_d
```

```python
for i in range(n):
    if sum_digit(nums[i]) == i:
        print(i)
        break
else:
    print(-1)
```
