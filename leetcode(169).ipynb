169. Majority Element

```python
nums = [2,2,1,1,1,2,2]
n = len(nums)
max_count = n//2
freq_count = {}

for num in nums:
    if num in freq_count:
        freq_count[num]+=1
    else:
        freq_count[num]=1
    
max_num = 0
for count in freq_count:
    if freq_count[count]>max_count:
        max_num = count

max_num
        
```
