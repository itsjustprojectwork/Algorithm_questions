```python
print('LEetcode Question 88: merge sorted array')
```

```python
arr1 = [-1,0,0,3,3,3,0,0,0]
arr2 = [1,2,2]
m = 6 #len(arr1)
n = 3 #len(arr2)
total_len = m+n
```

```python
j = 0
for i in range(m,total_len):
    arr1[i] = arr2[j]
    j+=1

for _ in range(total_len):
    for i in range(total_len-1):
        if arr1[i] > arr1[i+1]:
            arr1[i],arr1[i+1] = arr1[i+1],arr1[i]
    
print(arr1)
```
