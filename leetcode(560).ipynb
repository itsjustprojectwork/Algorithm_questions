560. Subarray Sum Equals K
<br>
Given an array of integers nums and an integer k, return the total number of subarrays whose sum equals to k.
<br>
A subarray is a contiguous non-empty sequence of elements within an array.
<br>
 

Example 1:<br>

Input: nums = [1,1,1], k = 2<br><br>
Output: 2<br>
Example 2:<br>

Input: nums = [1,2,3], k = 3<br>
Output: 2<br>
 

Constraints:<br>

1 <= nums.length <= 2 * 104<br>
-1000 <= nums[i] <= 1000<br>
-107 <= k <= 107<br>

```python
nums= [23,2,4,6,7]
k = 6
```

```python
count = 0
current_sum = 0
freq = {0:1}
for num in nums:
    current_sum +=num
    needed = current_sum - k
    if needed in freq:
        count+=freq[needed]
    if current_sum in freq:
        freq[current_sum]+=1
    else:
        freq[current_sum] = 1

count
        
```

```python
# subarrya
re = []
n = len(nums)
for i in range(n):
    sub = []
    for j in range(i,n):
        sub.append(nums[j])
        re.append(sub.copy())
re
```

```python
# TLE
n = len(nums)
count = 0
for i in range(n):
    current_sum = 0
    for j in range(i,n):
        current_sum+=nums[j]
        if current_sum == k:
            count+=1
count
```
