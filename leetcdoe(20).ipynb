20. Valid Parentheses
<br>
Given a string s containing just the characters '(', ')', '{', '}', '[' and ']', determine if the input string is valid.
<br>
An input string is valid if:
<br>
Open brackets must be closed by the same type of brackets.<br>
Open brackets must be closed in the correct order.<br>
Every close bracket has a corresponding open bracket of the same type.<br>
 <br>

Example 1:<br>
<br>
Input: s = "()"<br>
<br>
Output: true<br>
<br>
Example 2:<br>
<br>
Input: s = "()[]{}"<br>
<br>
Output: true<br>
<br>

Input: s = "(]"<br>

Output: false<br>

Example 4:<br>

Input: s = "([])"<br>

Output: true<br>

Example 5:<br>

Input: s = "([)]"<br>

Output: false<br>

```python
s = "([)]"
```

```python
# appraoch 1 
brackets = {
    '(':')',
    '[':']',
    '{':'}'
}
stack = []

for ch in s:
    if ch in brackets:
        stack.append(ch)
    elif ch in brackets.values():
        
        if ('(' in stack ) and ch == ')':
            stack.remove('(')
        elif ('[' in stack ) and ch == ']':
            stack.remove('[')
        elif ('{' in stack ) and ch == '}':
            stack.remove('{')
            
            
if len(stack)==0:
    print(True)
else:
    print(False)
```

```python
# appraoch 2
brackets = {
    ')':'(',
    ']':'[',
    '}':'{'
}
stack = []
for ch in s:
    if ch in '([{':
        stack.append(ch)
    else:
        if not stack or stack[-1]!=brackets[ch]:
            print(False)
            break
        stack.pop()
    
```
