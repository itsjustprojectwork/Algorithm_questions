50. Pow(x, n)
<br><br>
Implement pow(x, n), which calculates x raised to the power n (i.e., xn).
<br>
 

Example 1:
<br>
Input: x = 2.00000, n = 10<br>
Output: 1024.00000<br>
Example 2:<br>
<br>
Input: x = 2.10000, n = 3<br>
Output: 9.26100<br>
Example 3:<br>

Input: x = 2.00000, n = -2<br>
Output: 0.25000<br>
Explanation: 2-2 = 1/22 = 1/4 = 0.25<br>

complete

```python
# the perfect solution or the ideal solution cause its saving the repeated task
class Solution:
    def myPow(self, x: float, n: int) -> float:
            if n < 0:
                n = abs(n)

                if n % 2 != 0:
                    n = n - 1
                    half =  self.myPow(x, n // 2) 
                    ans = half * half * x 
                    return 1 / ans
                else:
                    half =  self.myPow(x, n // 2) 
                    ans = half * half 
                    return 1 / ans

            if n == 0:
                return 1

            if n == 1:
                return x * n

            if n % 2 != 0:
                n = n - 1
                half =  self.myPow(x, n // 2) 
                return half*half*x
            half =  self.myPow(x, n // 2) 
            return half*half
```
