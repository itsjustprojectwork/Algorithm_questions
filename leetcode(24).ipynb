24. Swap Nodes in Pairs
<br>
Given a linked list, swap every two adjacent nodes and return its head. You must solve the problem without modifying the values in the list's nodes (i.e., only nodes themselves may be changed.)
<br>
 
<br>
Example 1:
<br>
Input: head = [1,2,3,4]
<br>
Output: [2,1,4,3]
<br>
Explanation:
<br>
<br>

Example 2:
<br>
Input: head = []
<br>
Output: []
<br>
Example 3:
<br>
Input: head = [1]
<br>
Output: [1]
<br>
Example 4:
<br>
Input: head = [1,2,3]
<br>
Output: [2,1,3]
<br>
 

Constraints:
<br>
The number of nodes in the list is in the range [0, 100].<br>
0 <= Node.val <= 100<br>

```python
class Node:
    def __init__(self,val =0,next = None):
        self.val = val
        self.next = next
```

```python
values = [1,2,3,4,5]
head= None
tail = None
for val in values:
    node = Node(val)
    if head is None:
        head = node
        tail = node
    else:
        tail.next = node
        tail = node
```

```python
curr = head
while curr is not None:
    print(curr.val)
    curr = curr.next
```

```python
curr = head
prev = None
while curr is not None and curr.next is not None:
    first = curr
    second = curr.next
    first.next = second.next
    second.next = first
    
    if prev is None:
        head = second
    else:
        prev.next = second
    prev = first
    curr = first.next
    
```

```python
h = prev
while h is not None:
    print(h.val)
    h = h.next
```
