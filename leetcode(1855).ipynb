1855. Maximum Distance Between a Pair of Values
<br>
You are given two non-increasing 0-indexed integer arrays nums1​​​​​​ and nums2​​​​​​.
<br>
A pair of indices (i, j), where 0 <= i < nums1.length and 0 <= j < nums2.length, is valid if both i <= j and nums1[i] <= nums2[j]. <br>The distance of the pair is j - i​​​​.
<br>
Return the maximum distance of any valid pair (i, j). If there are no valid pairs, return 0.
<br>
An array arr is non-increasing if arr[i-1] >= arr[i] for every 1 <= i < arr.length.

 <br>

Example 1:<br>
<br>
Input: nums1 = [55,30,5,4,2], nums2 = [100,20,10,10,5]<br>
Output: 2<br>
Explanation: The valid pairs are (0,0), (2,2), (2,3), (2,4), (3,3), (3,4), and (4,4).<br>
The maximum distance is 2 with pair (2,4).<br>
Example 2:<br>
<br>
Input: nums1 = [2,2,2], nums2 = [10,10,1]<br>
Output: 1<br>
Explanation: The valid pairs are (0,0), (0,1), and (1,1).<br>
The maximum distance is 1 with pair (0,1).<br>
Example 3:
<br><br>
Input: nums1 = [30,29,19,5], nums2 = [25,25,25,25,25]<br>
Output: 2<br>
Explanation: The valid pairs are (2,2), (2,3), (2,4), (3,3), and (3,4).<br>
The maximum distance is 2 with pair (2,4).<br>


approach 1 using binary search

```python
nums1 = [30,29,19,5]
nums2 = [25,25,25,25,25]
 
n = len(nums1)
m = len(nums2)

max_dis = float('-inf')

for i in range(n):
    target = nums1[i]
    left = 0
    right = m-1
    can_idx = -1
    while left<=right:
        middle = (left+right) // 2
        if nums2[middle]>=target:
            if middle>=i:
                can_idx = middle
            left =middle + 1
        else:
            right = middle -1
            
    distance = can_idx - i
    if distance>max_dis:
        max_dis = distance
        
max_dis
        
    
    
```
