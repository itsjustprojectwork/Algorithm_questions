
70. Climbing Stairs
<br>
Hint
You are climbing a staircase. It takes n steps to reach the top.<br>
<br>
Each time you can either climb 1 or 2 steps. In how many distinct ways can you climb to the top?

```python
print('complete')
```

```python
def fibonnaci(n):
    if n ==1:
        return 1
    if n==2:
        return 2
    a =1
    b =2
    for i in range(2,n):
        nex = a+b
        a = b
        b= nex
    return nex
```

```python

print(fibonnaci(5))
```
