```python
nums = [120,12]

def reverse(n):
    rev = 0
    while n!=0:
        d = n%10
        rev = rev*10 +d
        n//=10
    return rev
```

```python
# approach 1. TLE

dis = float('inf')
n = len(nums)
for i in range(n):
    rev= reverse(nums[i])
    for j in range(i+1,n):
        if rev == nums[j]:
            print(nums[i],nums[j])
            x = abs(i-j)
            if x<dis:
                dis = x
                
dis
```

```python
arr = [9, 9]

freq = {}
n = len(arr)

for i in range(n):
    if arr[i] in freq:
        freq[arr[i]].append(i)
    else:
        freq[arr[i]] = [i]

min_dis = float('inf')

for i in range(n):
    x = reverse(arr[i])

    if x in freq:
        distances = freq[x]
        n = len(distances)
        target = i

        left = 0
        right = n - 1
        candidate = -1

        while left <= right:
            middle = (left + right) // 2

            if distances[middle] > target:
                candidate = distances[middle]
                right = middle - 1
            else:
                left = middle + 1

        if candidate != -1:
            distance = abs(target - candidate)

            if distance < min_dis:
                min_dis = distance

print(min_dis)
```
