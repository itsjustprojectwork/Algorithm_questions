```python
def freq_counter(s):
    freq= 0
    d = {}
    for ch in s:
        if ch in d:
            d[ch]+=1
        else:
            d[ch] = 1
        
    for count in d:
        if d[count]>1:
            freq +=1
    return freq
        
```

```python
s = "badc"
t ="baba"
map_st = {}
map_ts = {}
n = len(s)
for i in range(n):
    if s[i] in map_st:
        if map_st[s[i]] != t[i]:
            print(0)
            break
    if t[i] in map_ts:
        if map_ts[t[i]] != s[i]:
            print(0)
            break
    
    map_st[s[i]] = t[i]
    map_ts[t[i]] = s[i]
else:
    print(1)
    
```
