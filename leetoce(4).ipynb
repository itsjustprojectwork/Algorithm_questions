
4. Median of Two Sorted Arrays
<br>
Given two sorted arrays nums1 and nums2 of size m and n respectively, return the median of the two sorted arrays.
<br>
The overall run time complexity should be O(log (m+n)).

complete

```python
num1 = [1,2]
num2= [3,4]
n = len(num1)
m = len(num2)
total = m+n
total_nums = []
i = 0
j = 0
while i<n and j<m:
    if num1[i]<=num2[j]:
        total_nums.append(num1[i])
        i+=1
    else:
        total_nums.append(num2[j])
        j+=1

while i<n:
    total_nums.append(num1[i])
    i+=1
while j<m:
    total_nums.append(num2[j])
    j+=1

if total%2==0:
    idx = (total)//2
    median = (total_nums[idx]+ total_nums[idx-1]) / 2
else:
    idx = (total)//2
    median = float(total_nums[idx])
median
```
