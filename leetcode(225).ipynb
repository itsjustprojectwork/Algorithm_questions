225. Implement Stack using Queues
<br>
Implement a last-in-first-out (LIFO) stack using only two queues. The implemented stack should support all the functions of a normal stack (push, top, pop, and empty).

```python
print('complete')
```

```python
class Mystack:
    def __init__(self):
        self.nums = []
        
    def push(self,x)-> None:
        n= len(self.nums)
        self.nums.append(x)
        for i in range(n):
            self.nums[-(i + 1)], self.nums[-(i + 2)] = self.nums[-(i + 2)], self.nums[-(i + 1)]
        
    
    def pop(self)->int:
        x= self.nums[0]
        self.nums.remove(x)
        return x
    
    
    def top(self)->int:
        return self.nums[0]
    
    def empty(self)->bool:
        return len(self.nums)==0
```

```python
obj = Mystack()
obj.push(23)
obj.push(1)
print(obj.top())
```
