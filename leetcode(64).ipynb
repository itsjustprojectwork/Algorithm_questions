
64. Minimum Path Sum
<br>
Given a m x n grid filled with non-negative numbers, find a path from top left to bottom right, which minimizes the sum of all numbers along its path.
<br>
Note: You can only move either down or right at any point in time.<br>
Input: grid = [[1,3,1],[1,5,1],[4,2,1]]<br>
Output: 7<br>
Explanation: Because the path 1 → 3 → 1 → 1 → 1 minimizes the sum.<br>

complete
<br>
m == grid.length<br>
n == grid[i].length

```python
grid = [[1,2,3],[4,5,6]]
for g in grid:
    print(g)
```

```python
m= len(grid)-1
n = len(grid[0])-1
ans_grid = [[float('inf') for _ in row] for row in grid]
ans_grid[0][0]= grid[0][0]
for i in range(len(grid)):
    for j in range(len(grid[i])):
        if i==0 and j==0:
            continue
        ans_grid[i][j] = grid[i][j] + min(ans_grid[i][j-1],ans_grid[i-1][j])

print(ans_grid[m][n])
        
```

```python
def print_grid(grid):
    for g in grid:
        print(g)
```

```python
print_grid(ans_grid)
```
