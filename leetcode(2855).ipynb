2855. Minimum Right Shifts to Sort the Array
<br>
You are given a 0-indexed array nums of length n containing distinct positive integers. Return the minimum number of right shifts required to sort nums and -1 if this is not possible.
<br>
A right shift is defined as shifting the element at index i to index (i + 1) % n, for all indices.

 <br>

Example 1:<br>
<br>
Input: nums = [3,4,5,1,2]<br>
Output: 2<br>
Explanation: <br>
After the first right shift, nums = [2,3,4,5,1].<br>
After the second right shift, nums = [1,2,3,4,5].<br>
Now nums is sorted; therefore the answer is 2.<br>
Example 2:<br>
<br>
Input: nums = [1,3,5]<br>
Output: 0<br>
Explanation: nums is already sorted therefore, the answer is 0.<br>
Example 3:<br><br>

Input: nums = [2,1,4]<br>
Output: -1<br>
Explanation: It's impossible to sort the array using right shifts.<br>

```python
def check_sortation(nums):
    n = len(nums)
    for i in range(n-1):
        
        if nums[i]>nums[i+1]:
            return False
    else:
        return True
    
    
def right_shift(nums):
    n = len(nums)
    temp = nums[n-1]
    j = n-1
    while j!=0:
        nums[j] = nums[j-1]
        j-=1
    nums[0] = temp
    return nums
```

```python
nums = [6,5,1,2,3]
total = 0 
i = 0
j = len(nums)-1
while i<j:
    if nums[i]>nums[j]:
        total+=1
        right_shift(nums)
    else:
        break
    
    
if check_sortation(nums):
    print(total)
else:
    print(-1)
```
