206. Reverse Linked List
<br>
Given the head of a singly linked list, reverse the list, and return the reversed list.
<br>
 

Example 1:
<br>

Input: head = [1,2,3,4,5]<br>
Output: [5,4,3,2,1]<br>

```python
class Node:
    def __init__(self,val = 0,next=None):
        self.val = val
        self.next = next
        
```

```python
values = [1,2,3,4,5]
n = len(values)
head = None
tail = None
for i in range(n):
    node = Node(values[i])
    if head is None:
        head = node
        tail = node
    else:
        tail.next = node
        tail = node
```

```python
curr = head 
while curr!=None:
    print(curr.val)
    curr = curr.next
```

```python
def appraoch_1_using_rec(head):
    if head is None or head.next is None:
        return head
    ch = appraoch_1_using_rec(head.next)
    head.next.next = head
    head.next = None
    return ch

```

```python
cu = head
new_head = appraoch_1_using_rec(cu)

temp = new_head

while temp is not None:
    print(temp.val, end=" -> ")
    temp = temp.next
```

```python
def approach_2_reversing_using_recurison(head):
    if head is None or head.next is None:
        return (head,head)
    ch,tail = approach_2_reversing_using_recurison(head.next)
    tail.next = head
    head.next = None
    return ch,head
```

```python
curr = head 
while curr!=None:
    print(curr.val)
    curr = curr.next
```

```python
# approach 3 O(n)

curr = head
prev = None
while (curr is not None):
    temp = curr.next
    curr.next = prev
    prev = curr
    curr = temp
head = prev
```
