876. Middle of the Linked List
<br>
Given the head of a singly linked list, return the middle node of the linked list.
<br>
If there are two middle nodes, return the second middle node.
<br>
 

Example 1:

<br>
Input: head = [1,2,3,4,5]<br>
Output: [3,4,5]<br>
Explanation: The middle node of the list is node 3.<br>
Example 2:<br>


Input: head = [1,2,3,4,5,6]<br>
Output: [4,5,6]<br>
Explanation: Since the list has two middle nodes with values 3 and 4, we return the second one.<br>


<br><br>


in this type of problems we deal with the references and which are already stored they are not like list

```python
class Node:
    def __init__(self,val):
        self.val  = val
        self.next = None
```

```python
values=[2,5,6,7,89,9]
n = len(values)
head = None
tail = None
for i in range(n):
    node = Node(values[i])
    if head is None:
        head = node
        tail = node
    else:
        tail.next = node
        tail = node
```

```python
current = head
while current!=None:
    print(current.val)
    current = current.next
```

```python
fast = head
slow = head
while fast!=None and fast.next!=None:
    slow = slow.next
    fast = fast.next.next

print(slow)
```
