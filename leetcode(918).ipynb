Maximum Sum Circular Subarray<br><br>Given a circular integer array nums of length n, return the maximum possible sum of a non-empty subarray of nums.<br><br>A circular array means the end of the array connects to the beginning of the array. Formally, the next element of nums[i] is nums[(i + 1) % n] and the previous element of nums[i] is nums[(i - 1 + n) % n].<br><br>A subarray may only include each element of the fixed buffer nums at most once. Formally, for a subarray nums[i], nums[i + 1], ..., nums[j], there does not exist i <= k1, k2 <= j with k1 % n == k2 % n.<br><br>Example 1:<br>Input<br>nums = [1,-2,3,-2]<br>Output<br>3<br>Explanation<br>Subarray [3] has maximum sum 3.<br><br>Example 2:<br>Input<br>nums = [5,-3,5]<br>Output<br>10<br>Explanation<br>Subarray [5,5] has maximum sum 5 + 5 = 10.<br><br>Example 3:<br>Input<br>nums = [-3,-2,-3]<br>Output<br>-2<br>Explanation<br>Subarray [-2] has maximum sum -2.<br><br>Constraints:<br>- n == nums.length<br>- 1 <= n <= 3 * 104<br>- -3 * 104 <= nums[i] <= 3 * 104

```python
nums = [-3,-2,-3]
max_sum = float('-inf')
current_max= 0
current_min = 0
min_sum = float('inf')
total_sum = 0
for num in nums:
    # max_sum
    current_max+=num
    if current_max>num:
        current_max = current_max
    else:
        current_max = num
    if current_max>max_sum:
        max_sum = current_max
    
    # min_sum
    current_min+=num
    if current_min<num:
        current_min = current_min
    else:
        current_min = num
    if current_min<min_sum:
        min_sum = current_min
    total_sum+=num

if max_sum<0:
    print(max_sum)
else:
    print(max(max_sum,total_sum-min_sum))
```
