3512. Minimum Operations to Make Array Sum Divisible by K
<br>
You are given an integer array nums and an integer k. You can perform the following operation any number of times:
<br>
Select an index i and replace nums[i] with nums[i] - 1.<br>
Return the minimum number of operations required to make the sum of the array divisible by k.<br>

 

Example 1:<br>

Input: nums = [3,9,7], k = 5<br>
<br>
Output: 4<br>

Explanation:<br>

Perform 4 operations on nums[1] = 9. Now, nums = [3, 5, 7].<br>
The sum is 15, which is divisible by 5.<br>
Example 2:<br>
<br>
Input: nums = [4,1,3], k = 4<br>

Output: 0<br>

Explanation:<br>

The sum is 8, which is already divisible by 4. Hence, no operations are needed.<br>
Example 3:<br>

Input: nums = [3,2], k = 6<br>

Output: 5<br>

Explanation:<br>
<br>
Perform 3 operations on nums[0] = 3 and 2 operations on nums[1] = 2. Now, nums = [0, 0].
The sum is 0, which is divisible by 6.

```python
nums = [4,1,3] 
k = 4
s = 0
for num in nums:
    s+=num

s%k
```
