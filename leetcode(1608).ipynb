Special Array With X Elements Greater Than or Equal X<br><br>You are given an array nums of non-negative integers. nums is considered special if there exists a number x such that there are exactly x numbers in nums that are greater than or equal to x.<br><br>Notice that x does not have to be an element in nums.<br><br>Return x if the array is special, otherwise, return -1. It can be proven that if nums is special, the value for x is unique.<br><br>Example 1:<br>Input<br>nums = [3,5]<br>Output<br>2<br>Explanation<br>There are 2 values (3 and 5) that are greater than or equal to 2.<br><br>Example 2:<br>Input<br>nums = [0,0]<br>Output<br>-1<br>Explanation<br>No numbers fit the criteria for x.<br>If x = 0, there should be 0 numbers >= x, but there are 2.<br>If x = 1, there should be 1 number >= x, but there are 0.<br>If x = 2, there should be 2 numbers >= x, but there are 0.<br>x cannot be greater since there are only 2 numbers in nums.<br><br>Example 3:<br>Input<br>nums = [0,4,3,0,4]<br>Output<br>3<br>Explanation<br>There are 3 values that are greater than or equal to 3.<br><br>Constraints:<br>- 1 <= nums.length <= 100<br>- 0 <= nums[i] <= 1000

```python
nums = [3,5]
def count_num(nums,n):
    count = 0
    for num in nums:
        if num>=n:
            count+=1
    return count

i = 0
j = 10**9
while i<=j:
    mid = (i+j)//2
    val = count_num(nums,mid)
    if val > mid:
        i = mid +1
    elif val<mid:
        j = mid - 1
    else:
        print(val)
        break
else:
    print(-1)
    
```
