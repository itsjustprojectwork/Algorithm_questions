16. 3Sum Closest
<br>
Given an integer array nums of length n and an integer target, find three integers at distinct indices in nums such that the sum is closest to target.
<br>
Return the sum of the three integers.
<br>
You may assume that each input would have exactly one solution.
<br>
Example 1:
<br>
Input: nums = [-1,2,1,-4], target = 1<br>
Output: 2<br>
Explanation: The sum that is closest to the target is 2. (-1 + 2 + 1 = 2).<br>

complete


```python
nums = [10,20,30,40,50,60,90,100]
target = 1
nums.sort()
ans = abs(nums[0]+nums[1]+nums[2]-target)
ans_total = nums[0]+nums[1]+nums[2]
n = len(nums)
```

```python

for i in range(n-2):
    if i>0 and nums[i]==nums[i-1]:
        continue
    j = i+1
    k = n-1
    while j<k:
        total = nums[i]+nums[j]+nums[k]
        dis = abs(total - target)
        if dis<ans:
            ans = dis
            ans_total = total
        if total<target:
            j+=1
        elif total>target:
            k-=1
        else:
            
            j+=1
            k-=1
ans_total
    
```
