
48. Rotate Image
<br>
You are given an n x n 2D matrix representing an image, rotate the image by 90 degrees (clockwise).
<br>
You have to rotate the image in-place, which means you have to modify the input 2D matrix directly. DO NOT allocate another 2D matrix and do the rotation.

 <br>

Example 1:<br>
<br>

Input: matrix = [[1,2,3],[4,5,6],[7,8,9]]<br>
Output: [[7,4,1],[8,5,2],[9,6,3]]<br>
Example 2:<br>


Input: matrix = [[5,1,9,11],[2,4,8,10],[13,3,6,7],[15,14,12,16]]<br>
Output: [[15,13,2,5],[14,3,4,1],[12,6,8,9],[16,7,10,11]]<br>
 
<br>
Constraints:<br>

n == matrix.length == matrix[i].length<br>
1 <= n <= 20<br>
-1000 <= matrix[i][j] <= 1000<br>

```python
matrix = [[1,2,],[5,6]]


# appraoch 1 with using another 2D matrix
n = len(matrix)
m = len(matrix[0])
res = [[0]*m for _ in range(n)]
for i in range(n):
    for j in range(m):
        res[i][j] = matrix[n-1-j][i]
    
res
```

```python
n = len(matrix)

# reverse the columns
for i in range(n//2):
    for j in range(n):
        matrix[i][j],matrix[n-1-i][j] = matrix[n-1-i][j],matrix[i][j]
                
        # transpose where we do row into columns and columns into rows
for i in range(n):
    for j in range(i+1,n):
        matrix[i][j],matrix[j][i] = matrix[j][i],matrix[i][j]

matrix

        
```
