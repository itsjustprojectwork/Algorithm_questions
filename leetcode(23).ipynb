
23. Merge k Sorted Lists
<br>
You are given an array of k linked-lists lists, each linked-list is sorted in ascending order.
<br>
Merge all the linked-lists into one sorted linked-list and return it.
<br>
 
<br>
Example 1:
<br>
Input: lists = [[1,4,5],[1,3,4],[2,6]]<br>
Output: [1,1,2,3,4,4,5,6]<br>
Explanation: The linked-lists are:<br>
[
  1->4->5,<br>
  1->3->4,<br>
  2->6<br>
]
merging them into one sorted linked list:<br>
1->1->2->3->4->4->5->6<br>
Example 2:<br>
<br>
Input: lists = []<br>
Output: []<br>
Example 3:<br>
<br>
Input: lists = [[]]<br>
Output: []<br>

```python
class Node:
    def __init__(self,val = 0,next = None):
        self.val = val
        self.next = next
    
class opertaion_on_nodes:
    def fill_values_in_node(self,values):
        head = None
        tail = None
        for val in values:
            node  = Node(val)
            if head is None:
                head = node
                tail = node
            else:
                tail.next = node
                tail = node
        return head
    
    def print_nodes(self,head):
        if head is None:
            return None
        current = head
        while current!=None:
            print(current.val)
            current = current.next
```

```python
lists = [[1,4,5],[1,3,4],[2,6]]
operation = opertaion_on_nodes()
nodes_list = []
for list in lists:
    head = operation.fill_values_in_node(list)
    nodes_list.append(head)
```

```python
result = nodes_list[0]

def sort_merge(list1,list2):
    dummy = Node(0)
    t3 = dummy
    t1 = list1
    t2 = list2
    while t1!=None and t2!=None:
        if t1.val<t2.val:
            t3.next = t1
            t1 = t1.next
        else:
            t3.next = t2
            t2 = t2.next
            
        t3 = t3.next
    if t1 is not None:
        t3.next = t1
    else:
        t3.next = t2
    head = dummy.next
    return head

for head in nodes_list[1:]:
    result = sort_merge(result,head)
    

```

```python
operation.print_nodes(result)

    
```
