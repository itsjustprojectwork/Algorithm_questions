448. Find All Numbers Disappeared in an Array
<br>
Given an array nums of n integers where nums[i] is in the range [1, n], return an array of all the integers in the range [1, n] that do not appear in nums.
<br>
 

Example 1:<br>

Input: nums = [4,3,2,7,8,2,3,1]<br>
Output: [5,6]<br>
Example 2:<br>

Input: nums = [1,1]<br>
Output: [2]<br>
 

Constraints:<br>

n == nums.length<br>
1 <= n <= 105<br>
1 <= nums[i] <= n<br>

```python
nums = [4,3,2,7,8,2,3,1]
n = len(nums)+1
freq = {}
for num in nums:
    if num in freq:
        freq[num]+=1
    else:
        freq[num]=1
```

```python
for i in range(1,n):
    if i in freq:
        continue
    else:
        print(i)
```
