
1456. Maximum Number of Vowels in a Substring of Given Length
<br>
Given a string s and an integer k, return the maximum number of vowel letters in any substring of s with length k.
<br>
Vowel letters in English are 'a', 'e', 'i', 'o', and 'u'.
<br>
 
<br>
Example 1:<br>
<br>
Input: s = "abciiidef", k = 3<br>
Output: 3<br>
Explanation: The substring "iii" contains 3 vowel letters.<br>
Example 2:<br>
<br>
Input: s = "aeiou", k = 2<br>
Output: 2<br>
Explanation: Any substring of length 2 contains 2 vowels.<br>
Example 3:<br>
<br>
Input: s = "leetcode", k = 3<br>
Output: 2<br>
Explanation: "lee", "eet" and "ode" contain 2 vowels.<br>

```python
s = "abciiidef" 
k = 3
```

```python
# approach 1 fail tle

n = len(s) - k + 1
def vowels_count(s):
    vowels = ('a','e','i','o','u')
    count= 0
    for ch in s:
        if ch in vowels:
            count+=1
    return count

max_vowels = 0
for i in range(n):
    x = vowels_count(s[i:k+i])
    if  x>= max_vowels:
        max_vowels = x
        
max_vowels
```
