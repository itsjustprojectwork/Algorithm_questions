<b> in this question node its the refecence of the node in the linked list</b>

```python
class ListNode:
    def __init__(self,val=0,next =None):
        self.val = val
        self.next = next
        
```

```python
values = [1,2,3,4,5]

head = None
tail = None

for val in values:
    v = ListNode(val)
    if head is None:
      head= v
      tail = v
    else:
        tail.next = v
        tail = v

```
