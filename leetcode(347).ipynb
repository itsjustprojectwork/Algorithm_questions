
347. Top K Frequent Elements
<br>
Given an integer array nums and an integer k, return the k most frequent elements. You may return the answer in any order.
<br>
 <br>
<br>
Example 1:<br>

Input: nums = [1,1,1,2,2,3], k = 2<br>
<br>
Output: [1,2]<br>
<br>
Example 2:<br>
<br>
Input: nums = [1], k = 1<br>
<br><br>
Output: [1]<br>

Example 3:<br>

Input: nums = [1,2,1,2,1,2,3,1,3,2], k = 2<br>

Output: [1,2]<br>

 
<br>
Constraints:<br><br>
<br>
1 <= nums.length <= 105<br>
-104 <= nums[i] <= 104<br>
k is in the range [1, the number of unique elements in the array].<br>
It is guaranteed that the answer is unique.<br>

```python
nums = [2,2,3,1,1,1]
k = 2
```

```python
# this approach need O(n+m^2)

freq={}
for num in nums:
    if num in freq:
        freq[num]+=1
    else:
        freq[num]=1
ans = []
key = list(freq.keys())
n = len(key)

for i in range(n):
    for j in range(n-i-1):
        if freq[key[j]]<freq[key[j+1]]:
            key[j],key[j+1] = key[j+1],key[j]
key
        
```

```python
freq = {}
for num in nums:
    if num in freq:
        freq[num]+=1
    else:
        freq[num]=1
n = len(nums)
bucket = [] # ranking their index is ther frequency count and frequency cant be greater than len(nums)
for _ in range(n+1):
    bucket.append([])

for num in freq:
    bucket[freq[num]].append(num)

count = 0
ans = []
for i in range(n,0,-1):

    if len(bucket[i])>=1:
        for val in bucket[i]:
            ans.append(val)
            count+=1
            if count == k:
                break

    if count == k:
        break
ans
```
