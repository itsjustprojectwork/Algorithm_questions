A company intends to give its employees a pay rise.<br>

Write a solution to modify the salary column by multiplying each salary by 2.<br>

The result format is in the following example.<br>

 <br>

Example 1:<br>

Input:<br>
DataFrame employees<br>
+---------+--------+<br>
| name    | salary |<br>
+---------+--------+<br>
| Jack    | 19666  |<br>
| Piper   | 74754  |<br>
| Mia     | 62509  |<br>
| Ulysses | 54866  |<br>
+---------+--------+<br>
Output:<br>
+---------+--------+<br>
| name    | salary |<br>
+---------+--------+<br>
| Jack    | 39332  |<br>
| Piper   | 149508 |<br>
| Mia     | 125018 |<br>
| Ulysses | 109732 |<br>
+---------+--------+<br>
Explanation:<br>
Every salary has been doubled.<br>

```python
import pandas as pd

employees = pd.DataFrame({
    "name": ["Jack", "Piper", "Mia", "Ulysses"],
    "salary": [19666, 74754, 62509, 54866]
})

print(employees)
```

```python
employees['salary'] = employees['salary']*2
```
