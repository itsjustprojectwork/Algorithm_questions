203. Remove Linked List Elements
<br>
Given the head of a linked list and an integer val, remove all the nodes of the linked list that has Node.val == val, and return the new head.
<br>
 
<br>
Example 1:<br>
<br>

Input: head = [1,2,6,3,4,5,6], val = 6<br>
Output: [1,2,3,4,5]<br>
Example 2:<br>

Input: head = [], val = 1<br>
Output: []<br>
Example 3:<br>

Input: head = [7,7,7,7], val = 7
Output: []

```python
class Node:
    def __init__(self,val = 0 ,next = None):
        self.val = val
        self.next = next
        
        
```

```python
values = [2,2,3,4,5]
val = 2
n = len(values)
head = None
tail = None
for i in range(n):
    node= Node(values[i])
    if head is None:
        head = node
        tail = node
    else:
        tail.next = node
        tail  = node
```

```python
current = head
while current is not None:
    print(current.val)
    current = current.next
```

```python
c = head
prev = None
while c is not None:
    if c.val == val:
        if prev is None:
            head = c.next
        else:
            prev.next = c.next
    else:
        prev = c
    c = c.next
```
