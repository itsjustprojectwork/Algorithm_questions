1475. Final Prices With a Special Discount in a Shop<br><br>You are given an integer array prices where prices[i] is the price of the ith item in a shop.<br><br>There is a special discount for items in the shop. If you buy the ith item, then you will receive a discount equivalent to prices[j] where j is the minimum index such that j > i and prices[j] <= prices[i]. Otherwise, you will not receive any discount at all.<br><br>Return an integer array answer where answer[i] is the final price you will pay for the ith item of the shop, considering the special discount.<br><br>Example 1:<br>Input<br>prices = [8,4,6,2,3]<br>Output<br>[4,2,4,2,3]<br>Explanation<br>For item 0 with price[0]=8 you will receive a discount equivalent to prices[1]=4, therefore, the final price you will pay is 8 - 4 = 4.<br>For item 1 with price[1]=4 you will receive a discount equivalent to prices[3]=2, therefore, the final price you will pay is 4 - 2 = 2.<br>For item 2 with price[2]=6 you will receive a discount equivalent to prices[3]=2, therefore, the final price you will pay is 6 - 2 = 4.<br>For items 3 and 4 you will not receive any discount at all.<br><br>Example 2:<br>Input<br>prices = [1,2,3,4,5]<br>Output<br>[1,2,3,4,5]<br>Explanation<br>In this case, for all items, you will not receive any discount at all.<br><br>Example 3:<br>Input<br>prices = [10,1,1,6]<br>Output<br>[9,0,1,6]<br><br>Constraints:<br>- 1 <= prices.length <= 500<br>- 1 <= prices[i] <= 1000

```python
prices = [8,4,6,2,3]
```

```python
n = len(prices)
stack = []
ans = [-1]*n
for i in range(n-1,-1,-1):
    while stack:
        if prices[i]>=stack[-1]:
            ans[i] = prices[i] - stack[-1]
            break
        else:
            stack.pop()
    stack.append(prices[i])

for i in range(n):
    if ans[i] == -1:
        ans[i] = prices[i]

ans
```
