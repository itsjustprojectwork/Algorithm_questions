844. Backspace String Compare<br><br>Given two strings s and t, return true if they are equal when both are typed into empty text editors. '#' means a backspace character.<br><br>Note that after backspacing an empty text, the text will continue empty.<br><br>Example 1:<br>Input<br>s = "ab#c", t = "ad#c"<br>Output<br>true<br>Explanation<br>Both s and t become "ac".<br><br>Example 2:<br>Input<br>s = "ab##", t = "c#d#"<br>Output<br>true<br>Explanation<br>Both s and t become "".<br><br>Example 3:<br>Input<br>s = "a#c", t = "b"<br>Output<br>false<br>Explanation<br>s becomes "c" while t becomes "b".<br><br>Constraints:<br>- 1 <= s.length, t.length <= 200<br>- s and t only contain lowercase letters and '#' characters.<br>- Follow up: Can you solve it in O(n) time and O(1) space?

```python
s = "a##c"
t = "ad#c"
stack_s = []
for ch in s:
    if ch == '#':
        if not stack_s:
            continue
        else:
            stack_s.pop()
    else:
        stack_s.append(ch)
stack_t = []
for ch in t:
    if ch == '#':
        stack_t.pop()
    else:
        stack_t.append(ch)
        
''.join(stack_s) == ''.join(stack_t)
    
```
