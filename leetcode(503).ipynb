Next Greater Element II<br><br>Given a circular integer array nums (i.e., the next element of nums[nums.length - 1] is nums[0]), return the next greater number for every element in nums.<br><br>The next greater number of a number x is the first greater number to its traversing-order next in the array, which means you could search circularly to find its next greater number. If it doesn't exist, return -1 for this number.<br><br>Example 1:<br>Input<br>nums = [1,2,1]<br>Output<br>[2,-1,2]<br>Explanation<br>The first 1's next greater number is 2;<br>The number 2 can't find next greater number.<br>The second 1's next greater number needs to search circularly, which is also 2.<br><br>Example 2:<br>Input<br>nums = [1,2,3,4,3]<br>Output<br>[2,3,4,-1,4]<br><br>Constraints:<br>- 1 <= nums.length <= 104<br>- -109 <= nums[i] <= 109

```python
nums = [100, 1, 11, 1, 120, 111, 123, 1, -1, -100]
```

```python
# more readable and undestanding
stack = []
hashmap = {}
n = len(nums)

for i in range(n):
    while stack and stack[-1] < nums[i]:
        num = stack.pop()
        if num in hashmap:
            hashmap[num].append(i)
        else:
            hashmap[num] = [i]
    stack.append(nums[i])

for i in range(n):
    while (stack) and (stack[-1]< nums[i]):
        num = stack.pop()
        if num in hashmap:
            hashmap[num].append(i)
        else:
            hashmap[num]= [i]
ans = []
for num in nums:
    if num in hashmap:
          ans.append(nums[hashmap[num].pop(0)])
    else:
        ans.append(-1)
ans
```
