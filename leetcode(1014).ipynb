1014. Best Sightseeing Pair<br><br>You are given an integer array values where values[i] represents the value of the ith sightseeing spot. Two sightseeing spots i and j have a distance j - i between them.<br><br>The score of a pair (i < j) of sightseeing spots is values[i] + values[j] + i - j: the sum of the values of the sightseeing spots, minus the distance between them.<br><br>Return the maximum score of a pair of sightseeing spots.<br><br>Example 1:<br>Input<br>values = [8,1,5,2,6]<br>Output<br>11<br>Explanation<br>i = 0, j = 2, values[i] + values[j] + i - j = 8 + 5 + 0 - 2 = 11<br><br>Example 2:<br>Input<br>values = [1,2]<br>Output<br>2<br><br>Constraints:<br>- 2 <= values.length <= 5 * 104<br>- 1 <= values[i] <= 1000

```python
values = [8,1,5,2,6]
i =0
left_val = values[i]+ i
max_score = float('-inf')
for j in range(1,len(values)):
    left = values[i]+i
    if left>left_val:
        left_val = left
        
    right_val = values[j]-j
    score = left_val + right_val
    if score>max_score:
        max_score = score
    i+=1

max_score
```
