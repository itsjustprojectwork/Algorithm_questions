Peak Index in a Mountain Array<br><br>You are given an integer mountain array arr of length n where the values increase to a peak element and then decrease.<br><br>Return the index of the peak element.<br><br>Your task is to solve it in O(log(n)) time complexity.<br><br>Example 1:<br>Input<br>arr = [0,1,0]<br>Output<br>1<br><br>Example 2:<br>Input<br>arr = [0,2,1,0]<br>Output<br>1<br><br>Example 3:<br>Input<br>arr = [0,10,5,2]<br>Output<br>1<br><br>Constraints:<br>- 3 <= arr.length <= 105<br>- 0 <= arr[i] <= 106<br>- arr is guaranteed to be a mountain array.

```python
arr= [0,10,5,2]
i= 0
j = len(arr) - 1
while i<=j:
    mid = (i+j) // 2
    if arr[mid]<arr[mid+1]:
        i = mid + 1
    else:
        j = mid -1
i
```
