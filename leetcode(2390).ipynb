2390. Removing Stars From a String
<br>
You are given a string s, which contains stars *.
<br>
In one operation, you can:
<br>
Choose a star in s.<br>
Remove the closest non-star character to its left, as well as remove the star itself.<br>
Return the string after all stars have been removed.<br>

Note:<br>
<br>
The input will be generated such that the operation is always possible.<br>
It can be shown that the resulting string will always be unique.
 <br>
<br>
Example 1:<br>

Input: s = "leet**cod*e"<br>
Output: "lecoe"<br>
Explanation: Performing the removals from left to right:<br>
- The closest character to the 1st star is 't' in "leet**cod*e". s becomes "lee*cod*e".<br>
- The closest character to the 2nd star is 'e' in "lee*cod*e". s becomes "lecod*e".<br>
- The closest character to the 3rd star is 'd' in "lecod*e". s becomes "lecoe".<br>
There are no more stars, so we return "lecoe".
Example 2:<br>

Input: s = "erase*****"<br>
Output: ""<br>
Explanation: The entire string is removed, so we return an empty string.<br>
 
<br>
Constraints:
<br>
1 <= s.length <= 105<br>
s consists of lowercase English letters and stars *.<br>
The operation above can be performed on s.<br>

```python
s = "leet**cod*e"
```

```python
stack = []
for ch in s:
    if ch=='*':
        stack.pop()
    else:
        stack.append(ch)

stack
```

```python
''.join(stack)
```
