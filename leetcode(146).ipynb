146. LRU Cache
<br>
Design a data structure that follows the constraints of a Least Recently Used (LRU) cache.
<br>
Implement the LRUCache class:
<br>
LRUCache(int capacity) Initialize the LRU cache with positive size capacity.<br>
int get(int key) Return the value of the key if the key exists, otherwise return -1.<br>
void put(int key, int value) Update the value of the key if the key exists. Otherwise, add the key-value pair to the cache. If the <br>number of keys exceeds the capacity from this operation, evict the least recently used key.<br>
The functions get and put must each run in O(1) average time complexity.<br>

 <br>

```python
class LRUCache:
    def __init__(self, capacity):
        self.capacity = capacity
        self.stack = []
        self.store_val = {}
    def fix_priority(self,key):
        n = len(self.stack)
        for i in range(n):
            if self.stack[i] == key and i!=0:
                temp= self.stack[i]
                j = i
                while j!=0:
                    self.stack[j] = self.stack[j-1]
                    j-=1
                self.stack[0]= temp
                break
        return self.stack
    
    def put(self,key,val):
        if key in self.store_val:
            self.store_val[key]= val
        else:
            self.store_val[key] = val
    
        if key in self.stack:
            self.fix_priority(key)
        else:
            if len(self.stack)<self.capacity:
                self.stack.append(key)
                self.fix_priority(key)
            else:
                self.stack.pop()
                self.stack.append(key)
                self.fix_priority(key)
            
    def get(self,key):
        if key in self.stack:
            self.fix_priority(key)
            return self.store_val[key]
        else:
            return -1        
    
    def print_stack(self):
        return self.stack
```
