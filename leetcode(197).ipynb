```python
import pandas as pd

weather = pd.DataFrame({
    "id": [1, 2, 3, 4],
    "recordDate": [
        "2015-01-02",
        "2015-01-01",
        "2015-01-03",
        "2015-01-04"
    ],
    "temperature": [25, 10, 20, 30]
})

print(weather)
```

```python
weather.sort_values('recordDate',inplace= True)
weather
```

```python
weather['recordDate']  = pd.to_datetime(weather['recordDate'])
weather['next_day'] = weather['recordDate'] + pd.Timedelta(days=1)

```

```python
merger = weather.merge(weather,left_on='next_day',right_on = 'recordDate')
```

```python
merger[merger['temperature_x']<merger['temperature_y']][['id_y']].rename(columns = {'id_y':'id'})
```
