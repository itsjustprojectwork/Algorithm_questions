367. Valid Perfect Square<br><br>Given a positive integer num, return true if num is a perfect square or false otherwise.<br><br>A perfect square is an integer that is the square of an integer. In other words, it is the product of some integer with itself.<br><br>You must not use any built-in library function, such as sqrt.<br><br>Example 1:<br>Input<br>num = 16<br>Output<br>true<br>Explanation<br>We return true because 4 * 4 = 16 and 4 is an integer.<br><br>Example 2:<br>Input<br>num = 14<br>Output<br>false<br>Explanation<br>We return false because 3.742 * 3.742 = 14 and 3.742 is not an integer.<br><br>Constraints:<br>- 1 <= num <= 231 - 1

```python
n =  14
i = 0
j = n
while i<=j:
    mid = (i+j) // 2
    if mid * mid < n:
        i = mid +1
    elif mid * mid >n:
        j= mid- 1
    else:
        print(mid)
        break
else:
    print(False)
        
```
