3783. Mirror Distance of an Integer
<br>
You are given an integer n.
<br>
Define its mirror distance as: abs(n - reverse(n))​​​​​​​ where reverse(n) is the integer formed by reversing the digits of n.
<br>
Return an integer denoting the mirror distance of n​​​​​​​.
<br>
abs(x) denotes the absolute value of x.

 <br>

Example 1:
<br>
Input: n = 25
<br>
Output: 27
<br>
Explanation:
<br>
reverse(25) = 52.<br>
Thus, the answer is abs(25 - 52) = 27.<br>
Example 2:<br>
<br>
Input: n = 10<br>
<br>
Output: 9<br>
<br>
Explanation:<br>
<br>
reverse(10) = 01 which is 1.<br>
Thus, the answer is abs(10 - 1) = 9.<br>
Example 3:<br>
<br>
Input: n = 7<br>
<br>
Output: 0<br>
<br>
Explanation:<br>
<br>
reverse(7) = 7.<br>
Thus, the answer is abs(7 - 7) = 0.

```python
n = 21
origianl_num = n
revere = 0
while n!=0:
    digit = n%10
    revere = revere * 10 + digit
    n//=10
    
print(abs(origianl_num - revere))
```
