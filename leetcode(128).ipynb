128. Longest Consecutive Sequence
<br>
Given an unsorted array of integers nums, return the length of the longest consecutive elements sequence.
<br>
You must write an algorithm that runs in O(n) time.

 <br>

Example 1:
<br>
Input: nums = [100,4,200,1,3,2]<br>
Output: 4<br>
Explanation: The longest consecutive elements sequence is [1, 2, 3, 4]. Therefore its length is 4.<br>
Example 2:
<br>
Input: nums = [0,3,7,2,5,8,4,6,0,1]<br>
Output: 9<br>
Example 3:
<br>
Input: nums = [1,0,1,2]<br>
Output: 3<br>

complete

```python
nums = [100,4,200,1,3,2]

nums
```

```python
# brute force approach

n = len(nums)
for i in range(n):
    for j in range(n-i-1):
        if nums[j]>nums[j+1]:
            nums[j],nums[j+1] = nums[j+1],nums[j]
length = 1
for i in range(n-1):
    if nums[i]+1 == nums[i+1]:
        length+=1
length
```

```python
#O(n) approach cause set uses hasing to find element
exist = set(nums)
ans = 0
for num in exist:
    if num-1 not in exist:
        cur = num
        len = 1
        while cur+1 in exist:
            cur+=1
            len+=1
        
        ans = max(ans,len)
ans
```
