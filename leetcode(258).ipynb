258. Add Digits
<br>
Given an integer num, repeatedly add all its digits until the result has only one digit, and return it.

```python
def sum_digit(n):
    total = 0
    while n!=0:
        digit = n % 10
        total+=digit
        n//=10
    return total
```

```python
n =38
while n!=0:
    n= sum_digit(n)
    if n<10:
        print(n)
        break
    
```

```python
print(sum_digit(38))
```
