```python
# MY stack with size 10^3 
class Stack:
    def __init__(self):
        self.stack = [None] * 1000
        self.size = 0
        self.get_min = [None] * 1000
        self.min_size = 0
    
    def push(self,value):
        
        if self.size>=len(self.stack):
            raise OverflowError('Stack is fuckin over-flowing')
        # actual stack
        self.stack[self.size] = value
        self.size+=1
        
        
        # minimun stack values
        if self.min_size ==0:
            self.get_min[self.min_size] = value
            self.min_size+=1
        elif value<= self.get_min[self.min_size-1]:
            self.get_min[self.min_size] = value
            self.min_size+=1
    
    def pop(self):
        if self.size == 0:
            raise IndexError("mate its 0 you cant reduce it more DUMB!!")
        
        top_value = self.stack[self.size - 1]
        
        if top_value== self.get_min[self.min_size-1]:
            self.min_size -=1
            
        self.size-=1
        
    
    def top(self):
        if self.size ==0:
            raise IndexError('Bith thre are no elements!!')
        
        return self.stack[self.size - 1]
    
    def getMin(self):
        if self.min_size == 0:
            raise IndexError("Minimum of empty stack!!")

        return self.get_min[self.min_size - 1]
```

```python
# leetcode

class MinStack:

    def __init__(self):
        self.stack = []
        self.min_stack = []

    def push(self, value: int) -> None:
        self.stack.append(value)
        
        if self.min_stack:
            top_min = self.min_stack[-1]
            if value<=top_min:
                self.min_stack.append(value)
        else:
            self.min_stack.append(value)

    def pop(self) -> None:

        if self.stack[-1] == self.min_stack[-1]:
            self.min_stack.pop()

        self.stack.pop()

    def top(self) -> int:
        return self.stack[-1]

    def getMin(self) -> int:
        return self.min_stack[-1]
    

# Your MinStack object will be instantiated and called as such:
# obj = MinStack()
# obj.push(value)
# obj.pop()
# param_3 = obj.top()
# param_4 = obj.getMin()
```
