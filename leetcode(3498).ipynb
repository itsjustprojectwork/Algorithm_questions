3498. Reverse Degree of a String
<br>
Given a string s, calculate its reverse degree.
<br>
The reverse degree is calculated as follows:
<br>
For each character, multiply its position in the reversed alphabet ('a' = 26, 'b' = 25, ..., 'z' = 1) with its position in the string (1-indexed).<br>
Sum these products for all characters in the string.<br>
Return the reverse degree of s.<br>

 

Example 1:<br>
<br>
Input: s = "abc"<br>
<br>
Output: 148
<br>
Explanation:<br>

Letter	Index in Reversed Alphabet	Index in String	Product<br>
'a'	26	1	26<br>
'b'	25	2	50<br>
'c'	24	3	72<br>
The reversed degree is 26 + 50 + 72 = 148.<br>

Example 2:<br>
<br>
Input: s = "zaza"<br>

Output: 160<br>

Explanation:<br>

Letter	Index in Reversed Alphabet	Index in String	Product<br>
'z'	1	1	1<br>
'a'	26	2	52<br>
'z'	1	3	3<br>
'a'	26	4	104<br>
The reverse degree is 1 + 52 + 3 + 104 = 160.<br>

```python
class Solution:
    def reverseDegree(self, s: str) -> int:
        count = 1
        ans = 0
        for c in s:
            ans += count * (26 - (ord(c) - ord('a')))
            count+=1
        return ans
```
