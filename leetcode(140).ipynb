143. Reorder List
<br>
You are given the head of a singly linked-list. The list can be represented as:
<br>
L0 → L1 → … → Ln - 1 → Ln
Reorder the list to be on the following form:
<br>
L0 → Ln → L1 → Ln - 1 → L2 → Ln - 2 → …
You may not modify the values in the list's nodes. Only nodes themselves may be changed.
<br>
 <br>

Example 1:

<br>
Input: head = [1,2,3,4]<br>
Output: [1,4,2,3]<br>
Example 2:<br>
<br>
<br>
Input: head = [1,2,3,4,5]
Output: [1,5,2,4,3]

```python
class Node:
    def __init__(self,val =0,next = None):
        self.val = val
        self.next = next
        
```

```python
class Operations:
    def insert_nodes(self,values):
        head = None
        tail = None
        for val in values:
            node = Node(val)
            if head is None:
                head= node
                tail = node
            else:
                tail.next = node
                tail = node
        return head
        
    def print_nodes(self,head):
        if head is None:
            return f'fuckerrrr there are NOOO nodes'
        cur = head
        while cur!=None:
            print(cur.val)
            cur = cur.next
```

```python
operation = Operations()
head = operation.insert_nodes([1,2,3,4,5])
operation.print_nodes(head)
```

```python
fast = head
slow = head
last = None
while fast!=None and fast.next!=None:
    fast = fast.next.next
    slow = slow.next


# reverse the last past
rev = slow.next
prev = None
while rev!=None:
    temp = rev.next
    rev.next = prev
    prev = rev
    rev= temp
    

# seperate teh two half    
slow.next = None
    

# new merger alternative
dummy = Node(0)
new_head = dummy

cur = head

while cur!=None and prev!=None:
    cur_next = cur.next
    prev_next  = prev.next
    
    new_head.next = cur
    new_head = new_head.next
    
    
    new_head.next = prev
    new_head = new_head.next
    
    cur = cur_next
    prev = prev_next
    
if cur!=None:
    new_head.next = cur
    
```

```python
print('real_head:')
operation.print_nodes(head)
print()
print('reverse:')
operation.print_nodes(prev)
```
