 Duplicate Emails

```python
import pandas as pd
```

```python
emails = pd.DataFrame(
    {
       'id':[1,2,3,4],
       'email':['a@gmail.com','b@gmail.com','c@gmail.com','a@gmail.com'] 
    }
    )
```

```python
emails
```

```python
# appraoch 1
n = len(emails)
result = []
duplicates = []
for i in range(n):
    if  emails['email'][i] not in result:
        result.append(emails['email'][i])
    elif  emails['email'][i] not in duplicates:
        duplicates.append( emails['email'][i])
```

```python
count = {}
for email in emails['email']:
    if email in count:
        count[email]+=1
    else:
        count[email] = 1
    
dup = []

for email in count:
    if count[email]>1:
        dup.append(email)
        
dup
```
