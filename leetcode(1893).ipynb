1893. Check if All the Integers in a Range Are Covered
<br>
You are given a 2D integer array ranges and two integers left and right. Each ranges[i] = [starti, endi] represents an inclusive interval between starti and endi.
<br>
Return true if each integer in the inclusive range [left, right] is covered by at least one interval in ranges. Return false otherwise.
<br>
An integer x is covered by an interval ranges[i] = [starti, endi] if starti <= x <= endi.
<br>
 
<br>
Example 1:
<br>
Input: ranges = [[1,2],[3,4],[5,6]], left = 2, right = 5<br>
Output: true<br>
Explanation: Every integer between 2 and 5 is covered:<br>
- 2 is covered by the first range.<br>
- 3 and 4 are covered by the second range.<br>
- 5 is covered by the third range.<br>
Example 2:
<br>
Input: ranges = [[1,10],[10,20]], left = 21, right = 21<br>
Output: false<br>
Explanation: 21 is not covered by any range.<br>

```python
ranges =[[1,1]]

```

```python
left = 1
right = 50
```

```python
diff =[0]*52 # 52 cause last constaint of max endpoint is 50
for start,end in ranges:
    diff[start]+=1
    diff[end+1]-=1
    
coverage = 0
for i in range(1,52):
    coverage += diff[i]
    if left<=i and i<=right and coverage==0:
        print(False)
        break
else:
    print(True)
    
```
